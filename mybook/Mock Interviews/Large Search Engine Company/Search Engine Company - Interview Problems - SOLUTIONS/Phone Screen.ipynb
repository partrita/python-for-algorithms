## 기술적인 질문

**아래 *마크다운* 셀에서 다음 질문에 답하세요. 아래 셀에는 구문 강조 기능이 없다는 점에 유의하는 것이 중요합니다. 전화 화면 인터뷰에서 일반적으로 텍스트 편집기가 제공됩니다. 기본 텍스트 지원 외에는 아무것도 없습니다**

1. N번째 피보나치 수를 계산하는 함수를 작성하세요.

## 솔루션

이 질문에 대답하는 방법은 여러 가지가 있습니다. 여러 가지 방법으로 문제를 해결하고 각 방법의 장단점을 논의해야 할 수도 있습니다. 아래에는 다양한 솔루션이 나열되어 있습니다.

```python
## Example 1: Using looping technique
def fib(n):
    
    a,b = 1,1
    for i in range(n-1):
        a,b = b,a+b
    return a

print fib(7)
    
# Using recursion    
def fibR(n):
    if n==1 or n==2:
        return 1
    return fib(n-1)+fib(n-2)

print fibR(7)
 
## Example 3: Using generators
a,b = 0,1
def fibI():
    global a,b
    while True:
        a,b = b, a+b
        yield a
f=fibI()
f.next()
f.next()
f.next()
f.next()
f.next()
f.next()
print f.next()

 
## Example 4: Using memoization
def memoize(fn, arg):
    memo = {}
    if arg not in memo:
        memo[arg] = fn(arg)
    return memo[arg]
 
## fib() as written in example 1.
fibm = memoize(fib,7)
print fibm
 
## Example 5: Using memoization as decorator
class Memoize:
    def __init__(self, fn):
        self.fn = fn
        self.memo = {}
    def __call__(self, arg):
        if arg not in self.memo:
            self.memo[arg] = self.fn(arg)
            return self.memo[arg]
 
@Memoize
def fib(n):
    a,b = 1,1
    for i in range(n-1):
        a,b = b,a+b
    return a
print fib(7)
```
