# 해결책

5면 주사위는 한 번에 7가지 가능한 결과를 낼 수 없기 때문에 우리는 주사위를 최소한 두 번 굴려야 한다는 것을 즉시 알 수 있습니다.  

주사위를 두 번 굴리면 두 번 굴린 결과의 가능한 조합은 25가지가 됩니다. 25는 7로 나누어지지 않지만 21은 나누어집니다. 이는 의도한 범위에 없는 롤을 던지는 이전 전략을 구현할 수 있음을 의미합니다.

또한 버리는 것을 방지하기 위해 더 많은 롤을 구현하도록 솔루션을 확장할 수 없다는 점에 유의하는 것도 중요합니다. 왜냐하면 5와 7은 모두 소수이기 때문에 아무리 높이 올라가도 5의 지수는 7로 나누어지지 않는다는 뜻입니다.

우리는 가능한 25가지 롤 조합의 섹션으로 범위를 정의합니다. 이를 수행하는 좋은 방법은 두 개의 굴림을 1에서 25 사이의 고유한 결과 숫자로 변환하는 것입니다.

우리는 롤을 가져와 이 숫자를 생성한 다음 첫 번째 롤을 가져와서 1을 뺀 후 5를 곱합니다. 이제 첫 번째 롤은 0, 5, 10, 15 및 20 값에 해당합니다.

그런 다음 두 번째 롤을 가져와 첫 번째 조작 결과에 추가합니다. 1-25의 범위를 제공합니다.

그래서 우리의 최종 해결책은 주사위를 두 번 굴리는 것입니다. 조작 범위를 1부터 25까지 확인하고, 21보다 큰 경우 다시 굴립니다. 실제로 살펴보겠습니다.

```python
from random import randint


def dice5():
    return randint(1, 5)
```

이제 변환 기능을 살펴보겠습니다.

```python
def convert5to7():

    # For constant re-roll purposes
    while True:
        # Roll the dice twice
        roll_1 = dice5()
        roll_2 = dice5()

        # print 'The rolls were {} and {}'.format(roll_1,roll_2)

        # Convert the combination to the range 1 to 25
        num = ((roll_1 - 1) * 5) + (roll_2)

        # print 'The converted range number was:',num
        if num > 21:
            # re-roll if we are out of range
            continue

        return num % 7 + 1
```

```python
convert5to7()
```
