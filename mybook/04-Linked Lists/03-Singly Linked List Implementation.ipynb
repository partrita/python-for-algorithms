# 단일 연결 리스트 구현

이번 강의에서는 기본적인 Singly Linked List를 구현해보겠습니다.

단일 연결 목록에는 다른 노드에 대한 포인터가 있는 개별 노드로 항목의 순서가 지정된 목록이 있습니다.

```python
class Node(object):
    def __init__(self, value):

        self.value = value
        self.nextnode = None
```

이제 노드 컬렉션을 사용하여 Linked List를 구축할 수 있습니다.

```python
a = Node(1)
b = Node(2)
c = Node(3)
```

```python
a.nextnode = b
```

```python
b.nextnode = c
```

Linked List에서는 첫 번째 노드를head**라고 하고 마지막 노드를tail**이라고 합니다. Linked List의 장점과 단점을 살펴보겠습니다.

## 장점

* 연결 목록은 어느 위치에서나 일정한 시간 삽입 및 삭제를 수행하는 반면, 배열은 동일한 작업을 수행하는 데 O(n) 시간이 필요합니다.

* 연결 목록은 미리 크기를 지정하지 않고도 계속 확장될 수 있습니다. (배열 크기 조정에 대한 강의는 이 과정의 배열 순서 섹션에서 설명됩니다!)

## 단점

* 연결리스트의 요소에 접근하려면 목록의 선두부터 k번째 요소까지 이동하는 데 O(k) 시간이 필요합니다. 대조적으로, 배열은 배열의 요소에 액세스하기 위해 일정한 시간 작업을 수행합니다.
