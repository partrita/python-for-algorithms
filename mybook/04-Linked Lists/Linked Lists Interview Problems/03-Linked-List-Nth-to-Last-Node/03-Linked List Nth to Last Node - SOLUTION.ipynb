## 문제 설명
헤드 노드와 정수 값n**을 취한 다음 연결된 목록의 마지막 노드에서 n번째를 반환하는 함수를 작성하세요. 예를 들어, 다음과 같습니다.

```python
class Node:
    def __init__(self, value):
        self.value = value
        self.nextnode = None
```

```python
a = Node(1)
b = Node(2)
c = Node(3)
d = Node(4)
e = Node(5)

a.nextnode = b
b.nextnode = c
c.nextnode = d
d.nextnode = e

# This would return the node d with a value of 4, because its the 2nd to last node.
target_node = nth_to_last_node(2, a)
```

## 솔루션

이 문제에 대한 한 가지 접근 방식은 다음과 같습니다.

여러 노드와 n 노드 너비의 "블록"이 있다고 상상해 보십시오. 우리는 이 "블록"을 목록 아래로 끝까지 이동할 수 있으며, 일단 블록의 앞부분이 끝에 도달하면 블록의 다른 쪽 끝은 N번째 노드가 됩니다!  

따라서 이 "블록"을 구현하려면 왼쪽 및 오른쪽 포인터 쌍의 두 포인터만 있으면 됩니다. 우리가 취해야 할 단계를 표시해 보겠습니다.

* 헤드에서 하나의 포인터n노드를 이동합니다. 이것이 right_point가 됩니다.
* 다른 포인터를 머리 부분에 놓으면 이것이 left_point가 됩니다.
* 두 포인터 사이에 거리n**을 유지하면서 한 번에 한 노드씩 꼬리를 향해 블록(두 포인터)을 걷거나 횡단합니다.
* right_point가 꼬리에 닿으면 왼쪽 점이 목표에 있다는 것을 알 수 있습니다.

이에 대한 코드를 살펴보겠습니다!

```python
def nth_to_last_node(n, head):

    left_pointer = head
    right_pointer = head

    # Set right pointer at n nodes away from head
    for i in range(n - 1):
        # Check for edge case of not having enough nodes!
        if not right_pointer.nextnode:
            raise LookupError("Error: n is larger than the linked list.")

        # Otherwise, we can set the block
        right_pointer = right_pointer.nextnode

    # Move the block down the linked list
    while right_pointer.nextnode:
        left_pointer = left_pointer.nextnode
        right_pointer = right_pointer.nextnode

    # Now return left pointer, its at the nth to last element!
    return left_pointer
```
