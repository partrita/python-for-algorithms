# 단일 연결 목록 주기 확인 - 솔루션

## 문제

단일 연결 목록이 주어지면 단일 연결 목록의 첫 번째 노드를 가져와 연결 목록에 "주기"가 포함되어 있는지 나타내는 부울 값을 반환하는 함수를 작성하세요.

순환은 노드의 다음 지점이 실제로 목록의 이전 노드를 다시 가리키는 경우입니다. 이는 순환 연결 리스트라고도 합니다.

Linked List Node 클래스 코드가 제공되었습니다.

```python
class Node(object):
    def __init__(self, value):

        self.value = value
        self.nextnode = None
```

## 솔루션

이 문제를 해결하기 위해 목록을 통과하는 두 개의 마커가 있습니다.마커1및마커2**. 우리는 두 메이커 모두 목록의 첫 번째 노드에서 시작하여 연결 목록을 통과하도록 할 것입니다. 그러나 두 번째 마커인 marker2는 marker1이 이동하는 노드 1개마다 노드 2개를 앞으로 이동합니다.

이 논리를 통해 우리는 marker2가 더 빠르게 이동하면서 마커가 연결된 목록을 통해 "경주"하고 있다고 상상할 수 있습니다. 연결된 목록에 cylce가 있고 순환적으로 연결되어 있으면 트랙과 유사합니다. 이 경우 marker2는 결국 marker1을 "랩"하고 서로 동일하게 됩니다. 

연결된 목록에 순환이 없으면 marker2는 맨 처음 마커와 결코 같지 않고 끝까지 계속할 수 있어야 합니다.

이 논리가 코딩된 것을 살펴보겠습니다.

```python
def cycle_check(node):

    # Begin both markers at the first node
    marker1 = node
    marker2 = node

    # Go until end of list
    while marker2 != None and marker2.nextnode != None:
        # Note
        marker1 = marker1.nextnode
        marker2 = marker2.nextnode.nextnode

        # Check if the markers have matched
        if marker2 == marker1:
            return True

    # Case where marker ahead reaches the end of the list
    return False
```

# 솔루션 테스트

```python
"""
RUN THIS CELL TO TEST YOUR SOLUTION
"""
from nose.tools import assert_equal

# CREATE CYCLE LIST
a = Node(1)
b = Node(2)
c = Node(3)

a.nextnode = b
b.nextnode = c
c.nextnode = a  # Cycle Here!


# CREATE NON CYCLE LIST
x = Node(1)
y = Node(2)
z = Node(3)

x.nextnode = y
y.nextnode = z


#############
class TestCycleCheck(object):
    def test(self, sol):
        assert_equal(sol(a), True)
        assert_equal(sol(x), False)

        print("ALL TEST CASES PASSED")


# Run Tests

t = TestCycleCheck()
t.test(cycle_check)
```
