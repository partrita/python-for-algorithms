# 연결 목록 반전 - 솔루션

## 문제

Linked List를 제자리에 반전시키는 함수를 작성하세요. 이 함수는 목록의 헤드를 입력으로 사용하고 목록의 새 헤드를 반환합니다.

Linked List Node 클래스의 예가 제공됩니다.

```python
class Node(object):
    def __init__(self, value):

        self.value = value
        self.nextnode = None
```

# 솔루션

우리는 이 작업을 제자리에서 수행하고 싶기 때문에 기능이 O(1) 공간에서 작동하도록 만들고 싶습니다. 즉, 새 목록을 생성하고 싶지 않으므로 현재 노드를 사용하기만 하면 됩니다! 시간적으로는 O(n) 시간 안에 반전을 수행할 수 있습니다.

각 노드의 다음 포인터를 변경하여 목록을 뒤집을 수 있습니다. 각 노드의 다음 포인터는 이전 노드를 가리켜야 합니다.

입력 목록의 머리부터 끝까지 한 번만 통과하면 각 노드의 다음 포인터가 이전 요소를 가리키게 됩니다.

current.next_node를 이전으로 설정하기전에current.next_node를 next_node에 복사해야 합니다. 이 솔루션이 코딩된 것을 살펴보겠습니다.

```python
def reverse(head):

    # Set up current,previous, and next nodes
    current = head
    previous = None
    nextnode = None

    # until we have gone through to the end of the list
    while current:
        # Make sure to copy the current nodes next node to a variable next_node
        # Before overwriting as the previous node for reversal
        nextnode = current.nextnode

        # Reverse the pointer ot the next_node
        current.nextnode = previous

        # Go one forward in the list
        previous = current
        current = nextnode

    return previous
```

# 솔루션 테스트

자체 솔루션을 쉽게 테스트하여 작동하는지 확인할 수 있어야 합니다. 값이 1,2,3,4인 짧은 목록 a,b,c,d가 주어지면 역방향 함수의 효과를 확인하고 결과가 아래 논리와 일치하는지 확인하세요.

```python
# Create a list of 4 nodes
a = Node(1)
b = Node(2)
c = Node(3)
d = Node(4)

# Set up order a,b,c,d with values 1,2,3,4
a.nextnode = b
b.nextnode = c
c.nextnode = d
```

이제 a, b, c 뒤에 오는 노드의 값을 확인해 보겠습니다.

```python
print(a.nextnode.value)
print(b.nextnode.value)
print(c.nextnode.value)
```

```python
d.nextnode.value
```

지금까지는 너무 좋았습니다. 마지막 노드를 진행하는 데 가치가 없다는 점에 유의하십시오. 이는 의미가 있습니다! 이제 연결된 목록을 뒤집어 보겠습니다. 값의 순서가 반대인 것을 볼 수 있습니다!

```python
reverse(a)
```

```python
print(d.nextnode.value)
print(c.nextnode.value)
print(b.nextnode.value)
```
