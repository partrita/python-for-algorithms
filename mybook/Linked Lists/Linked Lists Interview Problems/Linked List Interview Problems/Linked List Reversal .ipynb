# 솔루션 테스트

**참고: 이는 솔루션 테스트를 위한 일반적인 실행 셀이 아닙니다. 아래 설명을 주의 깊게 읽어 보십시오**

자체 솔루션을 쉽게 테스트하여 작동하는지 확인할 수 있어야 합니다. 값이 1,2,3,4인 짧은 목록 a,b,c,d가 주어지면 역방향 함수의 효과를 확인하고 결과가 아래 논리와 일치하는지 확인하세요.

```python
# Create a list of 4 nodes
a = Node(1)
b = Node(2)
c = Node(3)
d = Node(4)

# Set up order a,b,c,d with values 1,2,3,4
a.nextnode = b
b.nextnode = c
c.nextnode = d
```

이제 a, b, c 뒤에 오는 노드의 값을 확인해 보겠습니다.

```python
print a.nextnode.value
print b.nextnode.value
print c.nextnode.value
```

```python
d.nextnode.value
```
