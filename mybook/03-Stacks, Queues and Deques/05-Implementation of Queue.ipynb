## 큐 메소드 및 속성


자체 대기열 구현을 시작하기 전에 대기열의 속성과 메서드를 검토해 보겠습니다.

* Queue()는 비어 있는 새 대기열을 생성합니다. 매개변수가 필요하지 않으며 빈 대기열을 반환합니다.
* enqueue(item)은 대기열 뒤쪽에 새 항목을 추가합니다. 항목이 필요하며 아무것도 반환하지 않습니다.
* dequeue()는 대기열에서 맨 앞의 항목을 제거합니다. 매개변수가 필요하지 않으며 항목을 반환합니다. 대기열이 수정되었습니다.
* isEmpty()는 대기열이 비어 있는지 테스트합니다. 매개변수가 필요하지 않으며 부울 값을 반환합니다.
* size()는 대기열에 있는 항목 수를 반환합니다. 매개변수가 필요하지 않으며 정수를 반환합니다.

____
## 대기열 구현

```python
class Queue:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def enqueue(self, item):
        self.items.insert(0, item)

    def dequeue(self):
        return self.items.pop()

    def size(self):
        return len(self.items)
```

```python
q = Queue()
```

```python
q.size()
```

```python
q.isEmpty()
```

```python
q.enqueue(1)
```

```python
q.dequeue()
```
