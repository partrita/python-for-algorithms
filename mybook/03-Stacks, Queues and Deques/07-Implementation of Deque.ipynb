# 데크 구현

이번 강의에서는 Deque 클래스를 직접 구현해보겠습니다!

## 메소드 및 속성

* Deque()는 비어 있는 새 deque를 생성합니다. 매개변수가 필요하지 않으며 빈 deque를 반환합니다.
* addFront(item)은 데크 앞에 새 항목을 추가합니다. 항목이 필요하며 아무것도 반환하지 않습니다.
* addRear(item)은 deque의 뒤쪽에 새 항목을 추가합니다. 항목이 필요하며 아무것도 반환하지 않습니다.
* RemoveFront()는 데크에서 앞 항목을 제거합니다. 매개변수가 필요하지 않으며 항목을 반환합니다. 데크가 수정되었습니다.
* RemoveRear()는 데크에서 뒤쪽 항목을 제거합니다. 매개변수가 필요하지 않으며 항목을 반환합니다. 데크가 수정되었습니다.
* isEmpty()는 데크가 비어 있는지 테스트합니다. 매개변수가 필요하지 않으며 부울 값을 반환합니다.
* size()는 deque의 항목 수를 반환합니다. 매개변수가 필요하지 않으며 정수를 반환합니다.

## 데크 구현

```python
class Deque:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def addFront(self, item):
        self.items.append(item)

    def addRear(self, item):
        self.items.insert(0, item)

    def removeFront(self):
        return self.items.pop()

    def removeRear(self):
        return self.items.pop(0)

    def size(self):
        return len(self.items)
```

```python
d = Deque()
```

```python
d.addFront("hello")
```

```python
d.addRear("world")
```

```python
d.size()
```

```python
print(d.removeFront() + " " + d.removeRear())
```

```python
d.size()
```
