# 스택 구현 -SOLUTION

**자세한 설명은 구현 강의를 참고하시기 바랍니다. 해당 강의의 코드는 다음과 같습니다.**

```python
class Stack(object):
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items) - 1]

    def size(self):
        return len(self.items)
```
