# 균형 괄호 확인 - 솔루션

## 문제 설명

여는 괄호와 닫는 괄호의 문자열이 주어지면 균형이 맞는지 확인하세요. 괄호에는 3가지 유형이 있습니다: 둥근 괄호: (), 대괄호: [], 중괄호: {}. 문자열에 이 문자 이외의 다른 문자가 포함되어 있지 않고 공백 단어나 숫자가 포함되어 있지 않다고 가정합니다. 참고로, 균형 잡힌 괄호를 사용하려면 모든 여는 괄호를 열린 순서의 역순으로 닫아야 합니다. 예를 들어 '([])'는 균형이 잡혀 있지만 '([)]'는 균형이 맞지 않습니다. 


입력 문자열에 공백이 없다고 가정할 수 있습니다.

## 솔루션

이것은 매우 일반적인 면접 질문이며 스택 사용에 대한 지식을 확인하는 주요 방법 중 하나입니다! 다음과 같이 솔루션 로직을 시작하겠습니다.

먼저 문자열을 왼쪽에서 오른쪽으로 스캔하고 여는 괄호를 볼 때마다 마지막 여는 괄호가 먼저 닫히기를 원하기 때문에 이를 스택에 푸시합니다. (스택의 FILO 구조를 기억하세요!)

그런 다음 닫는 괄호가 보이면 스택에서 요소를 팝하여 마지막으로 열린 괄호가 해당 닫는 일치인지 확인합니다. 유효한 일치이면 계속 진행하고 그렇지 않으면 false를 반환합니다. 

또는 스택이 비어 있으면 false를 반환합니다. 왜냐하면 이 닫는 괄호와 관련된 여는 괄호가 없기 때문입니다. 마지막으로 스택이 비어 있는지도 확인합니다. 그렇다면 true를 반환하고, 그렇지 않으면 닫히지 않은 열린 괄호가 있었기 때문에 false를 반환합니다. 

솔루션 예시는 다음과 같습니다.

```python
def balance_check(s):

    # Check is even number of brackets
    if len(s) % 2 != 0:
        return False

    # Set of opening brackets
    opening = set("([{")

    # Matching Pairs
    matches = set([("(", ")"), ("[", "]"), ("{", "}")])

    # Use a list as a "Stack"
    stack = []

    # Check every parenthesis in string
    for paren in s:
        # If its an opening, append it to list
        if paren in opening:
            stack.append(paren)

        else:
            # Check that there are parentheses in Stack
            if len(stack) == 0:
                return False

            # Check the last open parenthesis
            last_open = stack.pop()

            # Check if it has a closing match
            if (last_open, paren) not in matches:
                return False

    return len(stack) == 0
```

```python
balance_check("[]")
```

```python
balance_check("[](){([[[]]])}")
```

```python
balance_check("()(){]}")
```

# 솔루션 테스트

```python
"""
RUN THIS CELL TO TEST YOUR SOLUTION
"""
from nose.tools import assert_equal


class TestBalanceCheck(object):
    def test(self, sol):
        assert_equal(sol("[](){([[[]]])}("), False)
        assert_equal(sol("[{{{(())}}}]((()))"), True)
        assert_equal(sol("[[[]])]"), False)
        print("ALL TEST CASES PASSED")


# Run Tests

t = TestBalanceCheck()
t.test(balance_check)
```
