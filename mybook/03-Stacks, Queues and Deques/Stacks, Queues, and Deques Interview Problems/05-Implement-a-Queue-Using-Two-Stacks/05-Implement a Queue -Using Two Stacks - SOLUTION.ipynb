## 솔루션

핵심 통찰력은 스택이 순서를 바꾼다는 것입니다(큐는 그렇지 않습니다). 스택에 푸시된 일련의 요소는 팝될 때 반대 순서로 다시 나타납니다. 결과적으로, 서로 연결된 두 개의 스택은 동일한 순서로 요소를 반환합니다. 역순을 다시 역순으로 하는 것이 원래 순서이기 때문입니다.

 우리는 요소가 대기열에 추가될 때 채우는 in-stack을 사용하고 dequeue 작업은 out-stack에서 요소를 가져옵니다. 아웃 스택이 비어 있으면 인 스택의 모든 요소를 ​​팝하고 아웃 스택으로 푸시합니다.

```python
class Queue2Stacks(object):
    def __init__(self):

        # Two Stacks
        self.instack = []
        self.outstack = []

    def enqueue(self, element):

        # Add an enqueue with the "IN" stack
        self.instack.append(element)

    def dequeue(self):
        if not self.outstack:
            while self.instack:
                # Add the elements to the outstack to reverse the order when called
                self.outstack.append(self.instack.pop())
        return self.outstack.pop()
```

# 솔루션 테스트

스택과 큐에 대한 현재 지식을 바탕으로 이것이 제대로 작동하는지 알 수 있어야 합니다. 예를 들어 다음과 같이 인쇄되어야 합니다.

```python
"""
RUN THIS CELL TO CHECK THAT YOUR SOLUTION OUTPUT MAKES SENSE AND BEHAVES AS A QUEUE
"""
q = Queue2Stacks()

for i in range(5):
    q.enqueue(i)

for i in range(5):
    print(q.dequeue())
```
