# 스택 구현

## 스택 속성 및 메서드

자체 Stack 클래스를 구현하기 전에 Stack의 속성과 메서드를 검토해 보겠습니다.

스택 추상 데이터 유형은 다음 구조와 연산으로 정의됩니다. 위에서 설명한 대로 스택은 항목이 "상단"이라는 끝 부분에 추가되고 제거되는 정렬된 항목 모음으로 구성됩니다. 스택은 LIFO로 주문됩니다. 스택 작업은 다음과 같습니다.

* Stack()은 비어 있는 새 스택을 만듭니다. 매개변수가 필요하지 않으며 빈 스택을 반환합니다.
* push(item)은 스택의 맨 위에 새 항목을 추가합니다. 항목이 필요하며 아무것도 반환하지 않습니다.
* pop()은 스택의 맨 위 항목을 제거합니다. 매개변수가 필요하지 않으며 항목을 반환합니다. 스택이 수정되었습니다.
* peek()는 스택의 맨 위 항목을 반환하지만 제거하지는 않습니다. 매개변수가 필요하지 않습니다. 스택은 수정되지 않습니다.
* isEmpty()는 스택이 비어 있는지 테스트합니다. 매개변수가 필요하지 않으며 부울 값을 반환합니다.
* size()는 스택에 있는 항목 수를 반환합니다. 매개변수가 필요하지 않으며 정수를 반환합니다.

____

## 스택 구현

```python
class Stack:
    def __init__(self):
        self.items = []

    def isEmpty(self):
        return self.items == []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        return self.items.pop()

    def peek(self):
        return self.items[len(self.items) - 1]

    def size(self):
        return len(self.items)
```

한번 시험해 보자!

```python
s = Stack()
```

```python
print(s.isEmpty())
```

```python
s.push(1)
```

```python
s.push("two")
```

```python
s.peek()
```

```python
s.push(True)
```
