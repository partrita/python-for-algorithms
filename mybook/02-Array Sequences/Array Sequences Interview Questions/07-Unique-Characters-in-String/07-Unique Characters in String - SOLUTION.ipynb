# 문자열의 고유 문자

## 문제
문자열이 주어지면 그것이 모든 고유 문자로 구성되어 있는지 확인하십시오. 예를 들어 문자열 'abcde'는 모두 고유한 문자를 가지며 True를 반환해야 합니다. 문자열 'aabcde'에 중복 문자가 포함되어 있으므로 false를 반환해야 합니다.

## 솔루션
두 가지 가능한 솔루션을 보여 드리겠습니다. 하나는 내장된 데이터 구조와 내장된 함수를 사용하는 것이고, 다른 하나는 내장된 데이터 구조를 사용하지만 문자가 고유한지 확인하는 조회 방법을 사용하는 것입니다.

```python
def uni_char(s):
    return len(set(s)) == len(s)
```

```python
def uni_char2(s):
    chars = set()
    for let in s:
        # Check if in set
        if let in chars:
            return False
        else:
            # Add it to the set
            chars.add(let)
    return True
```
