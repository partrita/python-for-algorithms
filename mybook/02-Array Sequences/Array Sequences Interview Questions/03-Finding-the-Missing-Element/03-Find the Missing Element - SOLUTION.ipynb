# 누락된 요소 찾기

## 문제

음수가 아닌 정수의 배열을 고려하십시오. 두 번째 배열은 첫 번째 배열의 요소를 섞고 임의의 요소를 삭제하여 구성됩니다. 이 두 배열이 주어지면 두 번째 배열에서 어떤 요소가 누락되었는지 찾아보세요. 

다음은 입력 예입니다. 첫 번째 배열을 섞고 숫자 5를 제거하여 두 번째 배열을 구성합니다.

입력:
    
    파인더([1,2,3,4,5,6,7],[3,7,2,1,4,6])

출력:

    5는 누락된 숫자입니다.

## 솔루션

순진한 해결책은 두 번째 배열의 모든 요소를 살펴보고 첫 번째 배열에 나타나는지 확인하는 것입니다. 배열에 중복된 요소가 있을 수 있으므로 특별히 주의해야 합니다. 이 접근 방식의 복잡성은 O(N^2)입니다. 두 개의 for 루프가 필요하기 때문입니다.

더 효율적인 해결책은 첫 번째 배열을 정렬하는 것입니다. 따라서 첫 번째 배열의 요소가 두 번째 배열에 나타나는지 확인하는 동안 이진 검색을 수행할 수 있습니다(이진 검색에 대해서는 이후 섹션에서 자세히 알아볼 예정입니다). 하지만 우리는 여전히 중복 요소에 주의해야 합니다. 복잡도는 O(NlogN)입니다. 

중복된 숫자의 특별한 경우를 처리하고 싶지 않다면 두 배열을 모두 정렬하고 동시에 반복할 수 있습니다. 두 개의 반복자가 다른 값을 가지면 중지할 수 있습니다. 첫 번째 반복자의 값은 누락된 요소입니다. 이 해 역시 O(NlogN)입니다. 이 접근 방식에 대한 솔루션은 다음과 같습니다.

```python
def finder(arr1, arr2):

    # Sort the arrays
    arr1.sort()
    arr2.sort()

    # Compare elements in the sorted arrays
    for num1, num2 in zip(arr1, arr2):
        if num1 != num2:
            return num1

    # Otherwise return last element
    return arr1[-1]
```

```python
arr1 = [1, 2, 3, 4, 5, 6, 7]
arr2 = [3, 7, 2, 1, 4, 6]
finder(arr1, arr2)
```

대부분의 인터뷰에서는 선형 시간 솔루션을 제시할 것으로 예상됩니다. 해시테이블을 사용하여 각 요소가 두 번째 배열에 나타나는 횟수를 저장할 수 있습니다. 그런 다음 첫 번째 배열의 각 요소에 대해 카운터를 감소시킵니다. 개수가 0인 요소에 도달하면 해당 요소가 누락됩니다. 이 솔루션은 다음과 같습니다.

```python
import collections


def finder2(arr1, arr2):

    # Using default dict to avoid key errors
    d = collections.defaultdict(int)

    # Add a count for every instance in Array 1
    for num in arr2:
        d[num] += 1

    # Check if num not in dictionary
    for num in arr1:
        if d[num] == 0:
            return num

        # Otherwise, subtract a count
        else:
            d[num] -= 1
```

```python
arr1 = [5, 5, 7, 7]
arr2 = [5, 7, 7]

finder2(arr1, arr2)
```

한 가지 가능한 해결책은 arr1과 arr2의 모든 숫자의 합을 계산하고 array1의 합에서 arr2의 합을 빼는 것입니다. 차이점은 arr2에 숫자가 없다는 것입니다. 그러나 이 접근 방식은 배열이 너무 길거나 숫자가 매우 큰 경우 문제가 될 수 있습니다. 그러면 숫자를 합산하는 동안 오버플로가 발생합니다.

매우 영리한 트릭을 수행함으로써 선형 시간과 일정한 공간 복잡성을 문제 없이 달성할 수 있습니다. 변수를 0으로 초기화한 다음 해당 변수를 사용하여 첫 번째와 두 번째 배열의 모든 요소를 ​​[XOR](https://en.wikipedia.org/wiki/Exclusive_or)합니다. 결국 변수의 값은 array2의 요소가 누락된 결과입니다.

```python
def finder3(arr1, arr2):
    result = 0

    # Perform an XOR between the numbers in the arrays
    for num in arr1 + arr2:
        result ^= num
        print(result)

    return result
```

```python
finder3(arr1, arr2)
```
