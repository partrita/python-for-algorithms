# 배열 쌍 합계

## 문제

정수 배열이 주어지면 특정 값k**를 합산하는 *고유* 쌍을 모두 출력합니다.

따라서 입력은 다음과 같습니다.
    
    쌍_합([1,3,2,2],4)

**2쌍을 반환합니다.

     (1,3)
     (2,2)

**참고: 테스트 목적으로< 쌍 수를 출력하도록 기능을 변경하세요**

## 솔루션

O(N) 알고리즘은 집합 데이터 구조를 사용합니다. 처음부터 선형 패스를 수행하고 각 요소에 대해 k-요소가 표시된 숫자 집합에 있는지 확인합니다. 그렇다면 우리는 합 k 쌍을 찾아 출력에 추가합니다. 그렇지 않은 경우 이 요소는 아직 쌍에 속하지 않으며 표시된 요소 집합에 추가합니다. 

집합을 사용하는 방법을 알아내면 알고리즘은 정말 간단합니다. 복잡성은 O(N)입니다. 왜냐하면 우리는 배열의 단일 선형 스캔을 수행하고 각 요소에 대해 쌍을 형성하는 해당 숫자가 집합에 있는지 확인하거나 현재 요소를 집합에 추가하기 때문입니다. 집합의 삽입 및 찾기 작업은 모두 평균 O(1)이므로 알고리즘은 총 O(N)입니다.

```python
def pair_sum(arr, k):

    if len(arr) < 2:
        return

    # Sets for tracking
    seen = set()
    output = set()

    # For every number in array
    for num in arr:
        # Set target difference
        target = k - num

        # Add it to set if target hasn't been seen
        if target not in seen:
            seen.add(num)

        else:
            # Add a tuple with the corresponding pair
            output.add((min(num, target), max(num, target)))

    # FOR TESTING
    return len(output)
    # Nice one-liner for printing output
    # return '\n'.join(map(str,list(output)))
```

```python
pair_sum([1, 3, 2, 2], 4)
```

# 솔루션 테스트
