# 최대 연속 합계

## 문제
주어진 정수 배열(양수 및 음수)에서 가장 큰 연속 합계를 찾습니다. 

## 솔루션

배열이 모두 양수이면 결과는 단순히 모든 숫자의 합계입니다. 배열의 음수로 인해 시퀀스 확인을 시작해야 합니다.  

알고리즘은 숫자를 합산하고 현재 합계 변수에 저장하는 것입니다. 각 요소를 추가한 후 현재 합계가 지금까지 발생한 최대 합계보다 큰지 확인합니다. 그렇다면 최대 합계를 업데이트합니다. 현재 합계가 양수인 한 숫자를 계속 추가합니다. 현재 합계가 음수가 되면 새로운 현재 합계로 시작합니다. 음의 전류 합계는 미래 시퀀스의 합계만 감소시키기 때문입니다. 배열에는 모든 음의 정수가 포함될 수 있으므로 현재 합계를 0으로 재설정하지 않습니다. 그러면 결과는 가장 큰 음수가 됩니다. 

코드를 살펴보겠습니다:

```python
def large_cont_sum(arr):

    # Check to see if array is length 0
    if len(arr) == 0:
        return 0

    # Start the max and current sum at the first element
    max_sum = current_sum = arr[0]

    # For every element in array
    for num in arr[1:]:
        # Set current sum as the higher of the two
        current_sum = max(current_sum + num, num)

        # Set max as the higher between the currentSum and the current max
        max_sum = max(current_sum, max_sum)

    return max_sum
```

```python
large_cont_sum([1, 2, -1, 3, 4, 10, 10, -10, -1])
```
