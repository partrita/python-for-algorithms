# 철자 바꾸기 솔루션

## 문제

두 개의 문자열이 주어지면 그것이 철자법인지 확인하십시오. 철자 바꾸기는 두 문자열을 정확히 동일한 문자를 사용하여 쓸 수 있는 경우입니다(따라서 문자를 재배열하여 다른 문구나 단어를 얻을 수 있습니다). 

예를 들면:

    "홍보"는 "거짓말 위에 세워진 쓰레기"의 철자법입니다.
    
    클린트 이스트우드는 '올드 웨스트 액션'의 철자이다.
    
**참고: 공백과 대문자는 무시하세요. 따라서 "d go"는 "God", "dog" 및 "o d g"의 철자 바꾸기입니다.**

## 솔루션

이 문제에 대해 생각하는 방법에는 두 가지가 있습니다. 두 문자열의 문자/요소의 빈도가 동일한 경우(각 문자가 두 문자열에서 동일한 횟수로 표시됨을 의미) 서로의 철자 바꾸기입니다. 유사한 논리 방식에서 두 문자열이 정렬된 후 서로 동일하면 서로의 철자 바꾸기이기도 합니다.

이 두 번째 솔루션은 Python에서 매우 쉽게 구현할 수 있습니다.

```python
def anagram(s1, s2):

    # Remove spaces and lowercase letters
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()

    # Return boolean for sorted match.
    return sorted(s1) == sorted(s2)
```

```python
anagram("dog", "god")
```

```python
anagram("clint eastwood", "old west action")
```

```python
anagram("aa", "bb")
```

이제 위의 정렬 접근 방식은 간단하지만 실제로는 최적이 아니며 인터뷰 환경에서는 해시 테이블을 이해하는 능력을 테스트하기 위해 각 문자열의 문자 수를 세는 것과 관련된 보다 수동적인 솔루션을 구현하라는 요청을 받을 수도 있습니다. 계산 및 Python 사전을 사용하여 더 완전한 솔루션을 구축해 보겠습니다.

```python
def anagram2(s1, s2):

    # Remove spaces and lowercase letters
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()

    # Edge Case to check if same number of letters
    if len(s1) != len(s2):
        return False

    # Create counting dictionary (Note could use DefaultDict from Collections module)
    count = {}

    # Fill dictionary for first string (add counts)
    for letter in s1:
        if letter in count:
            count[letter] += 1
        else:
            count[letter] = 1

    # Fill dictionary for second string (subtract counts)
    for letter in s2:
        if letter in count:
            count[letter] -= 1
        else:
            count[letter] = 1

    # Check that all counts are 0
    for k in count:
        if count[k] != 0:
            return False

    # Otherwise they're anagrams
    return True
```

```python
anagram2("dog", "god")
```

```python
anagram2("clint eastwood", "old west action")
```

```python
anagram2("dd", "aa")
```

두 번째 솔루션에 대해 간략하게 설명하면 컬렉션 모듈에서 defaultdict를 사용하면 이 코드가 상당히 정리되고 최종 for 루프는 두 번째 for 루프에 내장될 수 있지만 위 구현에서는 모든 단계가 매우 명확합니다.
