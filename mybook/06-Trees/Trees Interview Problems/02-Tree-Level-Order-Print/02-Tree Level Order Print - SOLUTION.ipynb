## 솔루션

재귀를 사용하여 이 문제를 해결하는 것은 실용적이지 않습니다. 왜냐하면 재귀는 깊이 우선 검색과 유사하기 때문입니다. 그러나 여기서 필요한 것은 너비 우선 검색입니다. 따라서 이전에 너비 우선 검색에서 했던 것처럼 대기열을 사용하겠습니다. 먼저 루트 노드를 대기열에 푸시합니다. 그런 다음 조건 큐가 비어 있지 않은 상태에서 while 루프를 시작합니다. 그런 다음 각 반복마다 대기열의 시작 부분에서 노드를 팝하고 해당 자식을 대기열의 끝으로 푸시합니다. 노드를 팝하면 그 값과 공간을 인쇄합니다.

새 줄을 올바른 위치에 인쇄하려면 각 수준의 노드 수를 계산해야 합니다. 현재 레벨 카운트와 다음 레벨 카운트라는 2개의 카운트가 있습니다. 현재 레벨 개수는 새 줄을 인쇄하기 전에 이 레벨에서 인쇄해야 하는 노드 수를 나타냅니다. 대기열에서 요소를 팝하고 인쇄할 때마다 이를 감소시킵니다. 현재 레벨 수가 0에 도달하면 새 줄을 인쇄합니다. 다음 레벨 카운트에는 다음 레벨의 노드 수가 포함되며, 이는 새 줄을 인쇄한 후 현재 레벨 카운트가 됩니다. 현재 수준에 있는 노드의 자식 수를 계산하여 다음 수준의 노드 수를 계산합니다. 코드를 이해하는 것은 설명보다 쉽습니다.

```python
class Node:
    def __init__(self, val=None):
        self.left = None
        self.right = None
        self.val = val
```

```python
def levelOrderPrint(tree):
    if not tree:
        return
    nodes = collections.deque([tree])
    currentCount, nextCount = 1, 0
    while len(nodes) != 0:
        currentNode = nodes.popleft()
        currentCount -= 1
        print(
            currentNode.val,
        )
        if currentNode.left:
            nodes.append(currentNode.left)
            nextCount += 1
        if currentNode.right:
            nodes.append(currentNode.right)
            nextCount += 1
        if currentCount == 0:
            # finished printing current level
            print(
                "\n",
            )
            currentCount, nextCount = nextCount, currentCount
```

이 솔루션의 시간 복잡도는 트리의 노드 수인 O(N)이므로 최적입니다. 각 노드를 최소한 한 번씩 방문해야 하기 때문입니다. 공간 복잡도는 특정 지점에서 대기열의 최대 크기에 따라 달라집니다. 이는 한 수준에서 가장 많은 노드 수입니다. 최악의 경우는 트리가 완전한 이진 트리일 때 발생합니다. 즉, 각 수준이 가능한 최대 노드 수로 완전히 채워져 있음을 의미합니다. 이 경우 가장 많은 수의 노드가 마지막 수준에 나타나며 이는 (N+1)/2입니다. 여기서 N은 총 노드 수입니다. 따라서 공간 복잡도도 O(N)입니다. 대기열을 사용하는 동안에도 최적입니다.
