## 솔루션

우리는 트리의 사후 순회를 수행하여 이를 수행할 수 있습니다. 먼저 왼쪽 자식, 오른쪽 자식, 마지막으로 노드 자체를 처리합니다. 그래서 우리는 나뭇잎에서 시작하여 뿌리를 향해 새로운 트리를 아래쪽으로 형성합니다. 결과적으로 노드 자체를 처리하는 동안 왼쪽 및 오른쪽 하위 트리는 모두 유효한 트리밍 이진 검색 트리입니다(NULL일 수도 있음).

각 노드에서 해당 값을 기반으로 참조를 반환한 다음 현재 노드가 부모의 왼쪽 자식인지 오른쪽 자식인지에 따라 부모의 왼쪽 또는 오른쪽 자식 포인터에 할당됩니다. 현재 노드의 값이 min과 max 사이(min<=node<=max)이면 조치를 취할 필요가 없으므로 노드 자체에 대한 참조를 반환합니다. 현재 노드의 값이 min보다 작으면 오른쪽 하위 트리에 대한 참조를 반환하고 왼쪽 하위 트리를 삭제합니다. 노드의 값이 min보다 작으면 이진 검색 트리이기 때문에 왼쪽 자식은 확실히 min보다 작습니다. 하지만 오른쪽 자식은 min보다 작을 수도 있고 작을 수도 있고 확신할 수 없으므로 이에 대한 참조를 반환합니다. 상향식 후순 순회를 수행하고 있으므로 오른쪽 하위 트리는 이미 잘린 유효한 이진 검색 트리(NULL일 수 있음)이고 왼쪽 하위 트리는 확실히 NULL입니다. 왜냐하면 해당 노드는 확실히 min보다 작고 후순 순회 중에 제거되었기 때문입니다. 후위 순회에서는 먼저 노드의 모든 하위 항목을 처리한 다음 마지막으로 노드 자체를 처리한다는 점을 기억하십시오.

노드의 값이 max보다 클 때 비슷한 상황이 발생합니다. 이제 왼쪽 하위 트리에 대한 참조를 반환합니다. 노드의 값이 max보다 크면 오른쪽 자식도 확실히 max보다 크기 때문입니다. 그러나 왼쪽 자식은 최대값보다 클 수도 있고 그렇지 않을 수도 있습니다. 따라서 오른쪽 하위 트리를 버리고 이미 유효한 왼쪽 하위 트리에 대한 참조를 반환합니다. 코드는 이해하기 더 쉽습니다.

```python
def trimBST(tree, minVal, maxVal):

    if not tree:
        return

    tree.left = trimBST(tree.left, minVal, maxVal)
    tree.right = trimBST(tree.right, minVal, maxVal)

    if minVal <= tree.val <= maxVal:
        return tree

    if tree.val < minVal:
        return tree.right

    if tree.val > maxVal:
        return tree.left
```

이 알고리즘의 복잡도는 O(N)입니다. 여기서 N은 트리의 노드 수입니다. 기본적으로 우리는 모든 노드를 하나씩 방문하여 트리의 사후 순회를 수행하기 때문입니다. 모든 노드를 최소한 한 번씩 방문해야 하기 때문에 이는 최적입니다. 이것은 트리에서 재귀의 효율성을 보여주는 매우 우아한 질문입니다.
