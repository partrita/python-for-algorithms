# 너비 우선 검색 구현


Breath-First 검색이라는 대체 알고리즘은 DFS와 동일한 결과를 반환하는 기능을 제공하지만 최단 경로를 먼저 반환하도록 보장합니다. 이 알고리즘은 큐 데이터 구조를 사용하는 대신 재귀적 방식으로 구현하기가 조금 더 까다롭습니다. 따라서 반복적 접근 방식만 문서화하겠습니다. 탐색된 각 정점마다 수행되는 작업은 깊이 우선 구현과 동일하지만, 스택을 대기열로 대체하면 다음 단계로 이동하기 전에 정점 깊이의 폭을 탐색하게 됩니다. 이 동작은 발견된 첫 번째 경로가 비용 요소인 가장자리 수를 기반으로 존재하는 최단 경로 중 하나임을 보장합니다.

그래프의 형식은 다음과 같다고 가정합니다.

```python
graph = {
    "A": set(["B", "C"]),
    "B": set(["A", "D", "E"]),
    "C": set(["A", "F"]),
    "D": set(["B"]),
    "E": set(["B", "F"]),
    "F": set(["C", "E"]),
}
```

## 연결된 구성 요소
반복적인 DFS 구현과 마찬가지로 필요한 유일한 변경은 마지막 스택 대신 목록 구조의 시작 부분에서 다음 항목을 제거하는 것입니다.

```python
def bfs(graph, start):
    visited, queue = set(), [start]
    while queue:
        vertex = queue.pop(0)
        if vertex not in visited:
            visited.add(vertex)
            queue.extend(graph[vertex] - visited)
    return visited


bfs(graph, "A")
```

## 경로
이 구현은 두 정점 사이의 가능한 모든 경로를 반환하도록 약간 변경될 수 있으며, 그 중 첫 번째 정점은 가장 짧은 경로 중 하나입니다.

```python
def bfs_paths(graph, start, goal):
    queue = [(start, [start])]
    while queue:
        (vertex, path) = queue.pop(0)
        for next in graph[vertex] - set(path):
            if next == goal:
                yield path + [next]
            else:
                queue.append((next, path + [next]))


list(bfs_paths(graph, "A", "F"))
```

BFS 경로 생성기 메서드에서 최단 경로가 먼저 반환된다는 점을 알면 발견된 최단 경로를 반환하거나 경로가 없으면 '없음'을 반환하는 유용한 메서드를 만들 수 있습니다. 우리가 생성기를 사용하고 있기 때문에 이론적으로 이것은 BFS 구현에서 첫 번째 일치 경로를 분리하고 반환하는 것과 유사한 성능 결과를 제공해야 합니다.

```python
def shortest_path(graph, start, goal):
    try:
        return next(bfs_paths(graph, start, goal))
    except StopIteration:
        return None


shortest_path(graph, "A", "F")
```
