# 인접 목록으로 그래프 구현


사전을 사용하면 Python에서 인접 목록을 쉽게 구현할 수 있습니다. 그래프 추상 데이터 유형의 구현에서는 두 개의 클래스, 즉 정점의 마스터 목록을 보유하는Graph**와 그래프의 각 정점을 나타내는Vertex**를 생성합니다.

각 정점은 사전을 사용하여 연결된 정점과 각 가장자리의 가중치를 추적합니다. 이 사전은connectedTo**라고 합니다. 생성자는 일반적으로 문자열인 ID와connectedTo사전을 초기화합니다.addNeighbor메소드는 이 꼭지점에서 다른 꼭지점으로 연결을 추가하는 데 사용됩니다.getConnections메소드는connectedTo인스턴스 변수로 표시되는 인접 목록의 모든 정점을 반환합니다.getWeight메소드는 이 정점에서 매개변수로 전달된 정점까지의 가장자리 가중치를 반환합니다.

```python
class Vertex:
    def __init__(self, key):
        self.id = key
        self.connectedTo = {}

    def addNeighbor(self, nbr, weight=0):
        self.connectedTo[nbr] = weight

    def __str__(self):
        return str(self.id) + " connectedTo: " + str([x.id for x in self.connectedTo])

    def getConnections(self):
        return self.connectedTo.keys()

    def getId(self):
        return self.id

    def getWeight(self, nbr):
        return self.connectedTo[nbr]
```

그래프를 인접 목록으로 구현하기 위해 해야 할 일은 인접 목록 개체가 갖게 될 메서드를 정의하는 것입니다.

*Graph()**는 비어 있는 새 그래프를 생성합니다.
*addVertex(vert)**는 Vertex 인스턴스를 그래프에 추가합니다.
*addEdge(fromVert, toVert)두 꼭짓점을 연결하는 그래프에 방향이 있는 새로운 가장자리를 추가합니다.
*addEdge(fromVert, toVert, Weight)두 정점을 연결하는 그래프에 가중치가 부여되고 방향이 있는 새로운 가장자리를 추가합니다.
*getVertex(vertKey)**는 vertKey라는 그래프에서 정점을 찾습니다.
*getVertices()**는 그래프의 모든 정점 목록을 반환합니다. 
*in**은 vertex in graph 형식의 명령문에 대해 주어진 정점이 그래프에 있으면 True를 반환하고, 그렇지 않으면 False를 반환합니다.

```python
class Graph:
    def __init__(self):
        self.vertList = {}
        self.numVertices = 0

    def addVertex(self, key):
        self.numVertices = self.numVertices + 1
        newVertex = Vertex(key)
        self.vertList[key] = newVertex
        return newVertex

    def getVertex(self, n):
        if n in self.vertList:
            return self.vertList[n]
        else:
            return None

    def __contains__(self, n):
        return n in self.vertList

    def addEdge(self, f, t, cost=0):
        if f not in self.vertList:
            nv = self.addVertex(f)
        if t not in self.vertList:
            nv = self.addVertex(t)
        self.vertList[f].addNeighbor(self.vertList[t], cost)

    def getVertices(self):
        return self.vertList.keys()

    def __iter__(self):
        return iter(self.vertList.values())
```

이를 사용하는 방법에 대한 간단한 예를 살펴보겠습니다.

```python
g = Graph()
for i in range(6):
    g.addVertex(i)
```

```python
g.vertList
```

```python
g.addEdge(0, 1, 2)
```

```python
for vertex in g:
    print(vertex)
    print(vertex.getConnections())
    print("\n")
```
