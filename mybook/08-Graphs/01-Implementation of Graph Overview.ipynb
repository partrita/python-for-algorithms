# 그래프 개요 구현

이번 강의에서는 Node 클래스를 중심으로 간단한 그래프를 구현해보겠습니다. 면접 문제 해결 방법은 이 강의를 참고하세요

___
그래프는 방향이 지정되며 간선은 가중치를 가질 수 있습니다.

State 클래스, Node 클래스, 마지막으로 Graph 클래스라는 세 가지 클래스가 있습니다.

여기서는 두 가지 기본 제공 도구인 [OrderDict](https://docs.python.org/2/library/collections.html#collections.OrderedDict) 및 [Enum](https://docs.python.org/3/library/enum.html)을 활용하겠습니다.

```python
from enum import Enum


class State(Enum):
    unvisited = 1  # White
    visited = 2  # Black
    visiting = 3  # Gray
```

이제 Node 클래스의 경우 사전에 추가된 순서 키의 추적을 유지하려는 경우 OrderedDict 개체를 활용합니다.

```python
from collections import OrderedDict


class Node:
    def __init__(self, num):
        self.num = num
        self.visit_state = State.unvisited
        self.adjacent = OrderedDict()  # key = node, val = weight

    def __str__(self):
        return str(self.num)
```

그런 다음 마지막으로 그래프:

```python
class Graph:
    def __init__(self):
        self.nodes = OrderedDict()  # key = node id, val = node

    def add_node(self, num):
        node = Node(num)
        self.nodes[num] = node
        return node

    def add_edge(self, source, dest, weight=0):
        if source not in self.nodes:
            self.add_node(source)
        if dest not in self.nodes:
            self.add_node(dest)
        self.nodes[source].adjacent[self.nodes[dest]] = weight
```

```python
g = Graph()
g.add_edge(0, 1, 5)
```
