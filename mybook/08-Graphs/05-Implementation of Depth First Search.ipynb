# 깊이우선탐색 구현
우리가 논의할 이 알고리즘은 이름에서 알 수 있듯이 역추적하기 전에 각 분기 아래로 가능한 정점(제공된 루트에서)을 탐색하는 깊이 우선 검색입니다. 이 속성을 사용하면 알고리즘을 반복 및 재귀 형식 모두에서 간결하게 구현할 수 있습니다. 다음은 노드를 방문할 때마다 수행되는 작업 목록입니다.

* 현재 정점을 방문 중인 것으로 표시합니다.
* 방문 세트에 포함되지 않은 각 인접 정점을 탐색합니다.

우리는 다음과 같은 형식의 단순화된 그래프 버전을 가정합니다.

```python
graph = {
    "A": set(["B", "C"]),
    "B": set(["A", "D", "E"]),
    "C": set(["A", "F"]),
    "D": set(["B"]),
    "E": set(["B", "F"]),
    "F": set(["C", "E"]),
}
```

## 연결된 구성 요소

아래 구현에서는 스택 데이터 구조를 사용하여 주체 연결 구성 요소 내에서 액세스할 수 있는 정점 집합을 구축하고 반환합니다. 집합에서 항목을 제거하기 위해 Python의 빼기 연산자 오버로드를 사용하면 방문하지 않은 인접 정점만 추가할 수 있습니다.

```python
def dfs(graph, start):
    visited, stack = set(), [start]
    while stack:
        vertex = stack.pop()
        if vertex not in visited:
            visited.add(vertex)
            stack.extend(graph[vertex] - visited)
    return visited


dfs(graph, "A")
```

두 번째 구현은 첫 번째 구현과 동일한 기능을 제공하지만 이번에는 보다 간결한 재귀 형식을 사용합니다. 기본 매개변수 값이 한 번만 생성되는 일반적인 Python 문제로 인해 각 사용자 호출마다 새로운 방문 세트를 생성해야 합니다. 또 다른 Python 언어 세부 사항은 함수 변수가 참조로 전달되므로 각 재귀 호출 시 방문한 변경 가능 집합을 다시 할당할 필요가 없다는 것입니다.

```python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    for nxt in graph[start] - visited:
        dfs(graph, nxt, visited)
    return visited


dfs(graph, "A")
```

## 경로
이전 구현을 모두 조정하여 시작 정점과 목표 정점 사이의 가능한 모든 경로를 반환할 수 있습니다. 아래 구현에서는 스택 데이터 구조를 다시 사용하여 문제를 반복적으로 해결하고 목표를 찾을 때 가능한 각 경로를 생성합니다. 생성기를 사용하면 사용자는 원하는 양의 대체 경로만 계산할 수 있습니다.

```python
def dfs_paths(graph, start, goal):
    stack = [(start, [start])]
    while stack:
        (vertex, path) = stack.pop()
        for nxt in graph[vertex] - set(path):
            if nxt == goal:
                yield path + [nxt]
            else:
                stack.append((nxt, path + [nxt]))


list(dfs_paths(graph, "A", "F"))
```
