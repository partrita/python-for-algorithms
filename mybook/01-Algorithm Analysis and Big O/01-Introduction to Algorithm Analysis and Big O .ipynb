# 알고리즘 분석 및 Big O 소개

이번 강의에서는 알고리즘을 분석하는 방법과 그것이 왜 중요한지에 대해 논의하겠습니다!

## 알고리즘을 분석하는 이유는 무엇인가요?

시작하기 전에 알고리즘이 무엇인지 명확히합시다. 이 과정에서알고리즘**은 단순히 문제를 해결하기 위한 절차 또는 공식입니다. 일부 문제는 알고리즘에 이름이 있을 정도로 유명하며, 일부 절차는 이와 관련된 알고리즘에도 이름이 있을 정도로 일반적입니다. 이제 우리는 대답해야 할 좋은 질문이 있습니다.

*알고리즘을 어떻게 분석하고 알고리즘을 서로 비교할 수 있나요?*

당신과 친구가 모두 0부터 N까지의 숫자를 합하는 함수를 생각해냈다고 상상해 보십시오. 함수 내의 함수와 알고리즘을 어떻게 비교하시겠습니까? 두 사람 모두 다음 두 가지 별도의 기능을 생각해 냈다고 가정해 보겠습니다.

```python
# First function (Note the use of xrange since this is in Python 2)
def sum1(n):
    """
    Take an input of n and return the sum of the numbers from 0 to n
    """
    final_sum = 0
    for x in range(n + 1):
        final_sum += x

    return final_sum
```

```python
sum1(10)
```

```python
def sum2(n):
    """
    Take an input of n and return the sum of the numbers from 0 to n
    """
    return (n * (n + 1)) / 2
```

```python
sum2(10)
```

두 함수의 결과는 동일하지만 알고리즘은 완전히 다르다는 것을 알 수 있습니다. 첫 번째 함수는 반복적으로 숫자를 추가하는 반면 두 번째 함수는 다음을 사용합니다.
$$ \sum_{i=0}^{n} {i} = \frac{n(n+1)}{2} $$

그렇다면 어떻게 알고리즘을 객관적으로 비교할 수 있을까요? 메모리에서 차지하는 공간의 양을 비교하거나 각 기능을 실행하는 데 걸리는 시간을 비교할 수도 있습니다. jupyter에 내장된%timeit매직 함수를 사용하여 함수의 시간을 비교할 수 있습니다. Jupyter 노트북의 [**%timeit**](https://ipython.org/ipython-doc/3/interactive/magics.html#magic-timeit) 매직은 루프 반복을 특정 횟수만큼 반복하여 최상의 결과를 얻습니다. 설명서 링크를 확인하세요. 

계속해서 함수를 실행하는 데 걸린 시간을 비교해 보겠습니다.

```python
%timeit sum1(100)
```

```python
%timeit sum2(100)
```

두 번째 기능이 훨씬 더 효율적이라는 것을 알 수 있습니다! 처음보다 훨씬 빠른 속도로 달립니다. 그러나 "실행 시간"은 컴퓨터 자체의 속도와 하드웨어 성능에 따라 달라지기 때문에 객관적인 측정으로 사용할 수 없습니다. 따라서Big-O**라는 다른 방법을 사용해야 합니다!
