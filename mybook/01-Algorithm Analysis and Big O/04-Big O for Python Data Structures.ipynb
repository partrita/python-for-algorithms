# 파이썬 데이터 구조를 위한 Big O
이번 강의에서는 Python에 내장된 데이터 구조의 Big O인 목록과 사전에 대해 살펴보겠습니다.

# 목록

Python 목록에서는 동적 배열 역할을 하며 목록에서 호출되는 메서드를 통해 여러 가지 일반적인 작업을 지원합니다. 목록에서 수행되는 가장 일반적인 두 가지 작업은 인덱싱과 인덱스 위치에 할당하는 것입니다. 이러한 작업은 모두 일정한 시간 O(1)로 실행되도록 설계되었습니다.

[0,1,2...10000] 목록을 구성하기 위해 다양한 방법을 테스트하고 싶다고 가정해 보겠습니다. 목록 끝에 추가하기, 목록 연결하기, 캐스팅 및 목록 이해와 같은 도구 사용하기 등 다양한 방법을 비교해 보겠습니다.

예를 들어:

```python
def method1():
    l = []
    for n in range(10000):
        l = l + [n]


def method2():
    l = []
    for n in range(10000):
        l.append(n)


def method3():
    l = [n for n in range(10000)]


def method4():
    l = list(range(10000))
```

이제 timeit 매직 함수를 사용하여 이러한 메서드를 테스트해 보겠습니다.

```python
%timeit method1()
%timeit method2()
%timeit method3()
%timeit method4()
```

가장 효과적인 방법은 Python에 내장된 range() 함수라는 것을 분명히 알 수 있습니다!

효율적인 코드를 작성할 때 이러한 요소를 염두에 두는 것이 중요합니다. 더 중요한 것은 어떻게 O(1)로 인덱싱할 수 있는지 생각해 보는 것입니다. 이에 대해서는 일반적인 배열을 다룰 때 더 자세히 논의하겠습니다. 지금은 Big-O 효율성에 대한 개요를 아래 표에서 살펴보세요.
