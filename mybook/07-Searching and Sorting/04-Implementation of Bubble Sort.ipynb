# 버블 정렬 구현

버블 정렬은 목록을 여러 번 통과합니다. 인접한 품목을 비교하고, 고장난 품목을 교환합니다. 목록을 통과할 때마다 다음으로 큰 값이 적절한 위치에 배치됩니다. 본질적으로 각 항목은 해당 항목이 속한 위치까지 "거품"을 냅니다.

# 검토를 위한 리소스

버블 정렬에 대한 리뷰는 아래 리소스를 확인하세요!

* [위키피디아](https://en.wikipedia.org/wiki/Bubble_sort)
* [비주얼알고](http://visualgo.net/sorting.html)
* [애니메이션](http://www.cs.armstrong.edu/liang/animation/web/BubbleSort.html)
* [의사 코드를 사용한 정렬 알고리즘 애니메이션](http://www.sorting-algorithms.com/bubble-sort)

```python
def bubble_sort(arr):
    # For every element (arranged backwards)
    for n in range(len(arr) - 1, 0, -1):
        #
        for k in range(n):
            # If we come to a point to switch
            if arr[k] > arr[k + 1]:
                temp = arr[k]
                arr[k] = arr[k + 1]
                arr[k + 1] = temp
```

```python
arr = [3, 2, 13, 4, 6, 5, 7, 8, 1, 20]
bubble_sort(arr)
```
