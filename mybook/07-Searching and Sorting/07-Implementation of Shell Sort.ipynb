# 쉘 정렬 구현

쉘 정렬은 원래 목록을 여러 개의 작은 하위 목록으로 나누고 각 하위 목록은 삽입 정렬을 사용하여 정렬함으로써 삽입 정렬을 개선합니다. 이러한 하위 목록을 선택하는 독특한 방법은 셸 정렬의 핵심입니다. 목록을 인접한 항목의 하위 목록으로 나누는 대신, 쉘 정렬은 간격이라고 하는 증분 i를 사용하여 i 항목만큼 떨어져 있는 모든 항목을 선택하여 하위 목록을 만듭니다.

# 검토를 위한 리소스

Shell 정렬에 대한 검토는 아래 리소스를 확인하세요!

* [위키피디아](https://en.wikipedia.org/wiki/Shellsort)
* [비주얼알고](http://visualgo.net/sorting.html)
* [의사 코드를 사용한 정렬 알고리즘 애니메이션](http://www.sorting-algorithms.com/shell-sort)

```python
def shell_sort(arr):
    sublistcount = len(arr) // 2

    # While we still have sub lists
    while sublistcount > 0:
        for start in range(sublistcount):
            # Use a gap insertion
            gap_insertion_sort(arr, start, sublistcount)

        sublistcount = sublistcount // 2


def gap_insertion_sort(arr, start, gap):
    for i in range(start + gap, len(arr), gap):
        currentvalue = arr[i]
        position = i

        # Using the Gap
        while position >= gap and arr[position - gap] > currentvalue:
            arr[position] = arr[position - gap]
            position = position - gap

        # Set current value
        arr[position] = currentvalue
```

```python
arr = [45, 67, 23, 45, 21, 24, 7, 2, 6, 4, 90]
shell_sort(arr)
arr
```
