# 삽입정렬 구현

삽입 정렬은 한 번에 한 항목씩 최종 정렬된 배열(또는 목록)을 만듭니다. 퀵 정렬, 힙 정렬 또는 병합 정렬과 같은 고급 알고리즘보다 큰 목록에서는 효율성이 훨씬 떨어집니다.

# 검토를 위한 리소스

삽입 정렬에 대한 검토는 아래 리소스를 확인하세요!

* [위키피디아](https://en.wikipedia.org/wiki/Insertion_sort)
* [비주얼알고](http://visualgo.net/sorting.html)
* [애니메이션](http://cs.armstrong.edu/liang/animation/web/InsertionSort.html)
* [의사 코드를 사용한 정렬 알고리즘 애니메이션](http://www.sorting-algorithms.com/insertion-sort)

```python
def insertion_sort(arr):

    # For every index in array
    for i in range(1, len(arr)):
        # Set current values and position
        currentvalue = arr[i]
        position = i

        # Sorted Sublist
        while position > 0 and arr[position - 1] > currentvalue:
            arr[position] = arr[position - 1]
            position = position - 1

        arr[position] = currentvalue
```

```python
arr = [3, 5, 4, 6, 8, 1, 2, 12, 41, 25]
insertion_sort(arr)
arr
```
