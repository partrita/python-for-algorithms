# 퀵 정렬 구현

빠른 정렬에서는 먼저 피벗 값이라는 값을 선택합니다. 피벗 값을 선택하는 방법에는 여러 가지가 있지만 목록의 첫 번째 항목만 사용하겠습니다. 피벗 값의 역할은 목록 분할을 지원하는 것입니다. 일반적으로 분할점이라고 하는 최종 정렬 목록에서 피벗 값이 속하는 실제 위치는 빠른 정렬에 대한 후속 호출을 위해 목록을 나누는 데 사용됩니다.

# 검토를 위한 리소스

삽입 정렬에 대한 검토는 아래 리소스를 확인하세요!

* [위키피디아](https://en.wikipedia.org/wiki/Quicksort)
* [비주얼알고](http://visualgo.net/sorting.html)
* [의사 코드를 사용한 정렬 알고리즘 애니메이션](http://www.sorting-algorithms.com/quick-sort)

```python
def quick_sort(arr):

    quick_sort_help(arr, 0, len(arr) - 1)


def quick_sort_help(arr, first, last):

    if first < last:
        splitpoint = partition(arr, first, last)

        quick_sort_help(arr, first, splitpoint - 1)
        quick_sort_help(arr, splitpoint + 1, last)


def partition(arr, first, last):

    pivotvalue = arr[first]

    leftmark = first + 1
    rightmark = last

    done = False
    while not done:
        while leftmark <= rightmark and arr[leftmark] <= pivotvalue:
            leftmark = leftmark + 1

        while arr[rightmark] >= pivotvalue and rightmark >= leftmark:
            rightmark = rightmark - 1

        if rightmark < leftmark:
            done = True
        else:
            temp = arr[leftmark]
            arr[leftmark] = arr[rightmark]
            arr[rightmark] = temp

    temp = arr[first]
    arr[first] = arr[rightmark]
    arr[rightmark] = temp

    return rightmark
```

```python
arr = [2, 5, 4, 6, 7, 3, 1, 4, 12, 11]
quick_sort(arr)
arr
```
