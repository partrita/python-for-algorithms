____
## 순차검색

```python
def seq_search(arr, ele):
    """
    General Sequential Search. Works on Unordered lists.
    """

    # Start at position 0
    pos = 0
    # Target becomes true if ele is in the list
    found = False

    # go until end of list
    while pos < len(arr) and not found:
        # If match
        if arr[pos] == ele:
            found = True

        # Else move one down
        else:
            pos = pos + 1

    return found
```

```python
arr = [1, 9, 2, 8, 3, 4, 7, 5, 6]
```

```python
print(seq_search(arr, 1))
```

```python
print(seq_search(arr, 10))
```

## 주문 목록

목록이 다음보다 정렬되어 있다는 것을 안다면 요소나 그보다 큰 요소를 찾을 때까지만 확인하면 됩니다.

```python
def ordered_seq_search(arr, ele):
    """
    Sequential search for an Ordered list
    """
    # Start at position 0
    pos = 0

    # Target becomes true if ele is in the list
    found = False

    # Stop marker
    stopped = False

    # go until end of list
    while pos < len(arr) and not found and not stopped:
        # If match
        if arr[pos] == ele:
            found = True

        else:
            # Check if element is greater
            if arr[pos] > ele:
                stopped = True

            # Otherwise move on
            else:
                pos = pos + 1

    return found
```

```python
arr.sort()
```

```python
ordered_seq_search(arr, 3)
```

```python
ordered_seq_search(arr, 1.5)
```
