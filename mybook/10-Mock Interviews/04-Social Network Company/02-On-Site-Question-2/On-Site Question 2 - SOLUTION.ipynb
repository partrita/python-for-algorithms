# 현장질문 2 - SOLUTION

## 문제

중복이 포함된 계정 ID 번호(정수) 목록이 주어지면 하나의 고유한 정수를 찾습니다. (목록에는 하나의 고유한(중복되지 않은) 정수만 포함됨이 보장됩니다

## 요구사항

내장된 Python 함수나 메서드를 사용하지 마세요

## 솔루션

이는 이 과정의 배열 섹션에서 수행한 문제 중 하나와 매우 친숙하게 느껴질 것입니다! [XOR](https://en.wikipedia.org/wiki/Exclusive_or) 작업을 사용할 수 있습니다. **배타적 or** 연산은 두 개의 비트 세트를 사용하며, 각 쌍에 대해 비트 중 하나(둘 다는 아님)가 1인 경우 1 값을 반환합니다.

Python에서는 ^ 기호를 사용하여 XOR을 수행할 수 있습니다.

이제 우리의 솔루션에서는 목록의 모든 정수를 간단히 XOR할 수 있습니다. 0으로 설정된 고유 ID로 시작한 다음 목록에서 새 ID를 XOR할 때마다 비트가 변경됩니다. 동일한 ID를 다시 XOR하면 이전 변경 사항이 취소됩니다.

결국 우리는 단 한 번만 등장했던 고유한 ID를 갖게 될 것입니다!

```python
def solution(id_list):

    # Initiate unique Id
    unique_id = 0

    # XOR fo revery id in id list
    for i in id_list:
        # XOR operation
        unique_id ^= i

    return unique_id
```
