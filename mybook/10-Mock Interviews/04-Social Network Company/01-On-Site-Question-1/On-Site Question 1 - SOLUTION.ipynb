# 현장 질문 1 - 해결 방법

## 문제

정수 목록과 목표 숫자가 주어지면, 목표 숫자에 도달하기 위해 목록에서 두 개의 정수를 더할 수 있는지 여부를 나타내는 부울 값을 반환하는 함수를 작성하세요

## 요구사항

솔루션을 코딩하기 전에 펜이나 종이를 사용해 보십시오

정수 요소를 두 번 사용할 수 없습니다. 공간 대비 시간 최적화

## 솔루션

이 문제에서는set데이터 구조를 활용하겠습니다. 우리는 정수 목록을 단일 패스로 만들어 각 요소를 가능한 합계의 첫 번째 정수로 처리합니다.

각 반복마다 목표 숫자에 도달할 수 있는 두 번째 정수가 있는지 확인하고, 세트를 사용하여 목록에서 이미 본 적이 있는지 확인합니다.

그런 다음 반복의 현재 숫자를 추가하여 본 세트를 업데이트합니다.

이것이 코딩된 것을 봅시다:

```python
def solution(lst, target):

    # Create set to keep track of duplicates
    seen = set()

    # We want to find if there is a num2 that sums with num to reach the target

    for num in lst:
        num2 = target - num

        if num2 in seen:
            return True

        seen.add(num)

    # If we never find a pair match which creates the sum
    return False
```

```python
solution([1, 3, 5, 1, 7], 4)
```

```python
solution([1, 3, 5, 1, 7], 14)
```
