# 전화 화면 - 솔루션

## 문제 

첫 번째 항목만 유지하면서 주어진 문자열에서 중복 문자를 제거합니다. 예를 들어, 입력이 'tree traversal'이면 출력은 'tre avsl'이 됩니다.

## 솔루션

지금까지 본 문자를 추적하려면 효율적인 찾기 작업을 수행할 수 있는 데이터 구조가 필요합니다. 입력이 표준 ASCII 형식으로 보장되는 경우 크기 128의 부울 배열을 만들고 일정한 시간에 문자의 ASCII 값 인덱스에 액세스하여 조회를 수행할 수 있습니다. 그러나 문자열이 유니코드라면 100K보다 훨씬 큰 크기의 배열이 필요할 것입니다. 이는 대부분이 일반적으로 사용되지 않으므로 낭비가 됩니다.

데이터 구조 설정은 우리의 목적에 완벽하게 적합합니다. 키를 저장하고 키 존재에 대한 지속적인 검색을 제공합니다. 따라서 우리는 문자열의 문자를 반복하고 각 반복마다 세트를 검색하여 이전에 현재 문자를 본 적이 있는지 확인합니다. 세트에 있으면 이전에 본 적이 있다는 의미이므로 무시합니다. 그렇지 않은 경우 이를 결과에 포함하고 나중에 참조할 수 있도록 추적하기 위해 세트에 추가합니다. 코드는 이해하기 더 쉽습니다.

```python
def removeDuplicates(string):
    result = []
    seen = set()

    for char in string:
        if char not in seen:
            seen.add(char)
            result.append(char)

    return "".join(result)
```

알고리즘의 시간 복잡도는 O(N)입니다. 여기서 N은 입력 문자열의 문자 수입니다. set이 O(1) 삽입 및 찾기를 지원하기 때문입니다. 이는 가장 일반적인 문자열 인터뷰 질문 중 하나에 대한 최적의 솔루션입니다.
