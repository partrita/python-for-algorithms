## 솔루션

위 목록을 곱셈과 함께 보면 새 목록의 여러 항목에 대해 2 곱하기 3 또는 3 곱하기 4와 같이 곱셈이 반복된다는 것을 알 수 있습니다. 

우리는 욕심 많은 접근 방식을 취하고 나중에 다른 지수에서 재사용할 수 있도록 이러한 결과를 추적하기를 원할 것입니다. 또한 숫자가 0이면 어떻게 되는지 생각해야 합니다!

모든 정수의 곱(해당 인덱스의 정수 제외)을 찾기 위해 실제로 탐욕스러운 방식으로 목록을 두 번 살펴보겠습니다. 

첫 번째 패스에서는 각 인덱스이전**의 모든 정수의 곱을 구하고, 두 번째 패스에서는뒤로각 인덱스 뒤의 모든 정수의 곱을 구합니다.

그런 다음 최종 제품 답변을 얻으려면 각 인덱스 전후의 모든 제품을 곱하면 됩니다!

실제로 이를 살펴보겠습니다.

```python
def index_prod(lst):

    # Create an empty output list
    output = [None] * len(lst)

    # Set initial product and index for greedy run forward
    product = 1
    i = 0

    while i < len(lst):
        # Set index as cumulative product
        output[i] = product

        # Cumulative product
        product *= lst[i]

        # Move forward
        i += 1

    # Now for our Greedy run Backwards
    product = 1

    # Start index at last (taking into account index 0)
    i = len(lst) - 1

    # Until the beginning of the list
    while i >= 0:
        # Same operations as before, just backwards
        output[i] *= product
        product *= lst[i]
        i -= 1

    return output
```

```python
index_prod([1, 2, 3, 4])
```

```python
index_prod([0, 1, 2, 3, 4])
```

솔루션을 검토하고 이해했는지 확인하세요!  O(n) 시간과 O(n) 공간 복잡도를 사용합니다!
