# 전화 화면 

## 문제

**타워는 100층입니다. 당신은 계란 두 개를 받았습니다. 계란은 깨지지 않고 탑의 특정 층에서 떨어질 수 있을 만큼 강력합니다. 당신은 가능한 한 적은 수의 방울로 계란이 깨지지 않고 떨어질 수 있는 가장 높은 층을 찾는 임무를 받았습니다. 계란이 목표 층 위에서 떨어지면 계란이 깨집니다. 해당 층 이하에서 떨어뜨리면 그대로 유지되며 다른 층에 다시 달걀을 떨어뜨리는 테스트를 할 수 있습니다.**

**가능한 한 적은 방울로 이 작업을 수행하는 방법을 알고리즘으로 보여줍니다**
