# 현장 질문 1 - 해결 방법

## 문제

귀하는 Amazon 주식의 하루 동안의 과거 주가 목록을 받았습니다. 목록의 인덱스는 타임스탬프를 나타내므로 인덱스 0의 요소는 주식의 초기 가격이고, 인덱스 1의 요소는 해당 날짜에 기록된 다음 주식 가격입니다. 귀하의 임무는 해당 날짜에 Amazon 주식 단일주의 구매 및 판매에서 가능한 최대 이익을 반환하는 함수를 작성하는 것입니다. 이 작업을 최대한 효율적으로 수행하도록 노력하세요.**


예를 들어, 주가 목록이 주어졌다면:

가격 = [12,11,15,3,10]

그러면 함수는 가능한 최대 이익인 7을 반환합니다(3에 구매하고 10에 판매).

## 요구사항

IDE를 사용하지 않고 먼저 종이/연필로 이 문제를 해결해 보세요. 또한 가능한 모든 판매 조합을 무차별 대입하는 것보다 더 나은 솔루션을 생각해낼 수 있어야 한다는 점을 명심하세요

또한 주식을 "매도"할 수 없으며 주식을 매도하기 *전에* 매수해야 합니다.

## 솔루션

코딩을 시작하기 전에 몇 가지 사항에 대해 생각해 보겠습니다. 즉시 생각해 봐야 할 한 가지는 최대 가격과 최저 가격을 찾아 그 둘을 뺄 수 없다는 것입니다. 왜냐하면 최대 가격이 최소 가격보다 앞에 올 수 있기 때문입니다.

무차별 방식은 가능한 모든 가격 조합을 시도하는 것이지만 이는 O(N^2)이므로 매우 나쁩니다. 또한 이는 인터뷰 설정이므로 더 똑똑한 솔루션이 있다는 것을 이미 알고 계실 것입니다.

이 경우에는 [탐욕 알고리즘](https://en.wikipedia.org/wiki/Greedy_algorithm) 접근 방식을 사용합니다. 우리는 최대 이익을 추적하면서 주가 목록을 반복할 것입니다.

이는 모든 가격에 대해 지금까지 최저 가격을 추적한 다음 현재 최대 가격보다 더 나은 수익을 얻을 수 있는지 확인한다는 의미입니다.

이에 대한 구현을 살펴보겠습니다.

```python
def profit(stock_prices):

    # Start minimum price marker at first price
    min_stock_price = stock_prices[0]

    # Start off with a profit of zero
    max_profit = 0

    for price in stock_prices:
        # Check to set the lowest stock price so far
        min_stock_price = min(min_stock_price, price)

        # Check the current price against our minimum for a profit
        # comparison against the max_profit
        comparison_profit = price - min_stock_price

        # Compare against our max_profit so far
        max_profit = max(max_profit, comparison_profit)

    return max_profit
```

```python
profit([10, 12, 14, 12, 13, 11, 8, 7, 6, 13, 23, 45, 11, 10])
```

현재 우리는 한 패스 O(n)과 일정한 공간 O(1)에서 최대 이익을 찾고 있습니다. 그러나 우리는 여전히 극단적인 경우에 대해서는 생각하지 않습니다. 예를 들어 다음 시나리오를 해결해야 합니다.

* 주가는 항상 하락합니다.
* 목록에 주가가 2개 미만인 경우.

가격이 하루 종일 하락하면 마이너스 이익을 반환하여 첫 번째 시나리오를 처리할 수 있습니다(이렇게 하면 손실이 얼마나 되는지 알 수 있습니다). 두 번째 문제는 빠른len()확인을 통해 해결할 수 있습니다. 전체 솔루션을 살펴보겠습니다.

```python
def profit2(stock_prices):

    # Check length
    if len(stock_prices) < 2:
        raise Exception("Need at least two stock prices!")

    # Start minimum price marker at first price
    min_stock_price = stock_prices[0]

    # Start off with an initial max profit
    max_profit = stock_prices[1] - stock_prices[0]

    # Skip first index of 0
    for price in stock_prices[1:]:
        # NOTE THE REORDERING HERE DUE TO THE NEGATIVE PROFIT TRACKING

        # Check the current price against our minimum for a profit
        # comparison against the max_profit
        comparison_profit = price - min_stock_price

        # Compare against our max_profit so far
        max_profit = max(max_profit, comparison_profit)

        # Check to set the lowest stock price so far
        min_stock_price = min(min_stock_price, price)

    return max_profit
```

```python
# Exception Raised
profit2([1])
```

```python
profit2([30, 22, 21, 5])
```

엄청난! 이제 최악의 시나리오에 대비할 수 있습니다. 특히 원래 질문을 꽤 빨리 해결할 수 있는 경우에는 극단적인 경우를 염두에 두는 것이 중요합니다.
