___

**다음 질문에 답하세요(2~5분 응답). 기술적인 답변은 필요하지 않습니다. 귀하의 추론에 더 관심이 있습니다**

* 왜 여기서 일하고 싶나요?
* 가장 좋아하는 프로그래밍 언어는 무엇이며 그 이유는 무엇입니까?
* 5년 후 당신의 모습은?
