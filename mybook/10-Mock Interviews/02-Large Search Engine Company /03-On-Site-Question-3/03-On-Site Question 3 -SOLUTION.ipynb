# 해결책

이미 본 문제이므로 이 문제를 기억하시기 바랍니다. 해결책은 다음과 같습니다.

```python
def reverse(s):

    # Base Case
    if len(s) <= 1:
        return s

    # Recursion
    return reverse(s[1:]) + s[0]
```
