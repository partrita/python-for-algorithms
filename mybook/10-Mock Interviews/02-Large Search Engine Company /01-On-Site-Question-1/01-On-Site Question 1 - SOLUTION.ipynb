## 질문
1부터 7까지(균일한 확률로) 굴리는 주사위가 주어지면 5면 주사위를 시뮬레이션합니다. 가급적이면 솔루션을 함수로 작성하세요.

## 요구사항

반드시 펜과 종이 또는 화이트보드에 작성해야 합니다. 펜과 종이로 문제를 풀기 전까지는 실제 코딩이 허용되지 않습니다!

___

# 해결책

이것은 우리가 이전에 직접적으로 본 적이 없는 새로운 문제입니다! 여러 번 이 질문은 함수 형태로 요청됩니다. 주어진 함수 random_7()을 입력으로 가져와서 random_5()를 생성해야 합니다.

이 문제를 해결하는 열쇠는 롤의 최종 분포가 균일해야 한다는 요구 사항에 집중하는 것입니다. 또한 시간과 공간에 대한 요구 사항이 주어지지 않았으므로 솔루션은 실제로 *매우* 간단합니다. 5보다 큰 숫자를 얻으면 계속 다시 굴리기만 하면 됩니다!

우리는 이것을 코딩할 수 있습니다:

```python
from random import randint


def dice7():
    return randint(1, 7)


# Our Solution
def convert7to5():

    # Starting roll (just needs to be larger than 5)
    roll = 7

    while roll > 5:
        roll = dice7()
        print("dice7() produced a roll of ", roll)
    print(" Your final returned roll is below:")
    return roll
```

```python
convert7to5()
```

이제 다음 문제(현장 질문 2)는 더 어려워질 것입니다. 바로 롤의 역변환입니다! 이 질문은 질문에 대한 해결책을 지나치게 생각하지 않도록 상기시켜 주는 역할을 해야 합니다! 6과 7을 계속 굴리면 우리 솔루션이 무한대로 실행될 가능성이 있다는 점을 명심하십시오(그럴 가능성은 거의 없지만).
