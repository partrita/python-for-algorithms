# 현장질문 4 - SOLUTION

## 질문
**sqrt 함수를 사용하지 않고 주어진 숫자의 제곱근을 가장 가까운 정수로 내림하여 구합니다. 예를 들어, [9, 15] 사이 숫자의 제곱근은 3을 반환하고 [16, 24]는 4를 반환해야 합니다.**

## 요구사항

자유롭게 코딩하세요(그러나 종이/연필 또는 화이트보드를 사용하는 것이 좋습니다)**

## 솔루션

(음수가 아닌) 숫자 N의 제곱근은 항상 0과 N/2 사이에 있습니다. 이 문제를 해결하는 간단한 방법은 k의 제곱이 N보다 크거나 r이 될 때까지 0과 N/2 사이의 모든 숫자 k를 확인하는 것입니다. k^2가 N과 같게 되면 k를 반환합니다. 그렇지 않으면 반내림하므로 k-1을 반환합니다. 코드는 다음과 같습니다.

```python
def solution(num):
    if num < 0:
        raise ValueError
    if num == 1:
        return 1
    for k in range(1 + (num // 2)):
        if k**2 == num:
            return k
        elif k**2 > num:
            return k - 1
    return k
```

```python
solution(14)
```

```python
solution(15)
```

```python
solution(16)
```

이 접근 방식의 복잡성은 O(N)입니다. 최악의 경우 N/2 숫자를 확인해야 하기 때문입니다. 이 선형 알고리즘은 매우 비효율적입니다. 일종의 이진 검색을 사용하여 속도를 높일 수 있습니다. 결과가 0과 N/2 사이에 있다는 것을 알고 있으므로 먼저 N/4를 시도하여 그 제곱이 N보다 작은지, 크거나 같은지 확인할 수 있습니다. 같으면 해당 값을 반환합니다. 그보다 작으면 N/4와 N/2 사이에서 검색을 계속합니다. 그렇지 않고 더 크면 0과 N/4 사이에서 검색합니다. 두 경우 모두 잠재적 범위를 절반으로 줄이고 계속합니다. 이것이 이진 검색의 논리입니다. 하지만 우리는 일반적인 이진 검색을 수행하지 않고 수정되었습니다. 우리는 k^2<=N이지만 (k+1)^2>N인 숫자 k에서 멈추기를 원합니다. 예를 들어:

```python
def better_solution(num):
    if num < 0:
        raise ValueError
    if num == 1:
        return 1
    low = 0
    high = 1 + (num / 2)

    while low + 1 < high:
        mid = low + (high - low) / 2
        square = mid**2
        if square == num:
            return mid
        elif square < num:
            low = mid
        else:
            high = mid

    return low
```

```python
better_solution(14)
```

```python
better_solution(15)
```

```python
better_solution(16)
```

일반 이진 검색과의 한 가지 차이점은 while 루프의 조건입니다. 이는 low<high ​​대신 low+1<high입니다. 또한 low=mid+1 대신 low=mid를 사용하고 high=mid-1 대신 high=mid를 사용합니다. 이는 표준 이진 검색에 대한 수정 사항입니다. 복잡도는 여전히 동일하지만 로그 O(logN)입니다. 순진한 선형 솔루션보다 훨씬 낫습니다.

영리한 수학 트릭을 포함하는 상수 시간 O(1) 솔루션도 있습니다. 여기 있습니다:

# $$ \sqrt{N} = N^{0.5} = 2^{\log_2 N^{0.5}} = 2^{0.5 \log_2 N} $$

이 솔루션은 숫자의 로그 지수를 취하면 결과가 변하지 않고 여전히 숫자 자체라는 속성을 활용합니다. 따라서 먼저 숫자의 로그를 계산하고 0.5를 곱한 다음 지수를 구하고 마지막으로 해당 값의 하한선을 취하여 반올림할 수 있습니다. 이렇게 하면 log 함수를 사용하여 sqrt 함수 사용을 피할 수 있습니다. 가장 가까운 정수로 내림된 숫자의 로그는 숫자의 이진 표현에서 가장 왼쪽 1의 위치를 ​​보면 상수 시간에 계산될 수 있습니다. 예를 들어, 이진수 6은 110이고 가장 왼쪽의 1은 위치 2에 있습니다(오른쪽부터 0부터 시작). 따라서 반올림된 6의 로그는 2입니다. 하지만 이 솔루션은 반올림 효과로 인해 항상 위의 알고리즘과 동일한 결과를 제공하지는 않습니다. 그리고 면접관의 관점에 따라 이 접근 방식은 매우 우아하고 영리하거나 까다롭고 유효하지 않은 것으로 간주될 수 있습니다. 어느 쪽이든, 당신은 지름길을 할 수 있다는 것을 면접관에게 알려야 합니다!
