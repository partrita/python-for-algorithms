# 현장 질문 1 - 해결 방법

## 문제

주어진 정수 목록에서 목록에 있는 3개의 정수로 만들 수 있는 가장 큰 제품을 찾으세요

## 요구사항

목록에는 항상 최소 3개의 정수가 있다고 가정할 수 있습니다

종이/연필만 사용하세요. 손으로 최대한 풀기 전까지는 코딩하지 마세요.

## 솔루션

우리는 O(1) 공간을 사용하여 O(n) 시간에 이 문제를 해결할 수 있으며 음수도 고려할 수 있으므로 [-5,-5,1,3]과 같은 목록은 (-5)(-5)(3) = 75를 답으로 반환합니다.

여러분이 이 문제와 전자상거래 회사 모의 면접 질문에서 나온 아마존 주식 문제 사이의 유사성을 깨닫기 시작하셨기를 바랍니다! 단순히 세 자리 숫자의 모든 조합을 시도하여 이 문제를 무차별 대입 공격으로 해결할 수 있지만 이 경우 O(n^3) 시간이 필요합니다!

욕심 많은 접근 방식을 사용하여 몇 가지 숫자를 추적하는 것은 어떻습니까? 주식 문제에서는 지금까지 최대 이익을 추적했지만, 이 문제에서는 실제로 여러 숫자를 추적할 것입니다.

* 현재까지 3개 수치 중 최고치
* 현재까지 2개 수치 중 최고치
* 현재까지 가장 높은 수치

음수를 고려하고 싶기 때문에 2의 가장 낮은 곱과 가장 낮은 숫자도 추적합니다.

* 2중 가장 낮은 제품
* 가장 낮은 숫자

목록을 반복하고 끝에 도달하면 3개의 숫자로 가능한 가장 높은 제품을 얻게 됩니다. 각 반복에서 우리는 현재 가장 높은 곱인 3을 가져와 현재 정수에 가장 높은 곱과 가장 낮은 곱인 2를 곱한 것과 비교합니다.

이것이 코딩된 것을 봅시다:

```python
def solution(lst):

    # Start at index 2 (3rd element) and assign highest and lowest
    # based off of first two elements

    # Highest Number so far
    high = max(lst[0], lst[1])

    # Lowest number so far
    low = min(lst[0], lst[1])

    # Initiate Highest and lowest products of two numbers
    high_prod2 = lst[0] * lst[1]
    low_prod2 = lst[0] * lst[1]

    # Initiate highest product of 3 numbers
    high_prod3 = lst[0] * lst[1] * lst[2]

    # Iterate through list
    for num in lst[2:]:
        # Compare possible highest product of 3 numbers
        high_prod3 = max(high_prod3, num * high_prod2, num * low_prod2)

        # Check for possible new highest products of 2 numbers
        high_prod2 = max(high_prod2, num * high, num * low)

        # Check for possible new lowest products of 2 numbers
        low_prod2 = min(low_prod2, num * high, num * low)

        # Check for new possible high
        high = max(high, num)

        # Check for new possible low
        low = min(low, num)

    return high_prod3
```

```python
l = [
    99,
    -82,
    82,
    40,
    75,
    -24,
    39,
    -82,
    5,
    30,
    -25,
    -94,
    93,
    -23,
    48,
    50,
    49,
    -81,
    41,
    63,
]

solution(l)
```

엄청난! 탐욕적 접근 방식을 사용하여 O(n) 시간 내에 문제를 완료할 수 있었습니다. 목록을 반복해야 하고 무차별 대입 솔루션이 기하급수적으로 진행될 때 이러한 종류의 접근 방식을 염두에 두십시오!
