# 전화 화면 - 솔루션

## 문제

n개의 정수 목록이 주어졌고 그 목록이 정렬되어 있다는 것을 알고 있다면 주어진 정수가 목록에 있는지 얼마나 빨리 확인할 수 있습니까? 일반적인 추론 및 검색 방법에 대해 자세히 설명**

## 요구사항

다른 사람에게 귀하의 솔루션을 설명하고 그것이 이해되는지 확인하십시오. 이 문제에 대해서는 아무 것도 코딩하지 마세요

## 솔루션

이 문제가 친숙하게 들렸으면 좋겠습니다! 목록이 이미 정렬되어 있으므로 이진 검색을 사용하여 정수를 검색할 수 있습니다! 즉, [O(logn) 시간과 O(1) 공간](http://bigochheatsheet.com/)에서 항목을 찾을 수 있습니다!
