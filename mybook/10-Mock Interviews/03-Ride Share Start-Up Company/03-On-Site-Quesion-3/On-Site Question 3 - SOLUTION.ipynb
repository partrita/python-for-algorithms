## 솔루션

마음에 떠오르는 첫 번째 해결책은 모든 노드에서 그 값이 왼쪽 자식보다 크거나 같고 오른쪽 자식보다 작거나 같은지 확인하는 것입니다(같음이 왼쪽이나 오른쪽에 나타날 수 있다고 가정). 그러나 이 접근 방식은 노드가 조부모 또는 조상의 조건을 위반하는지 여부를 확인하지 않기 때문에 잘못되었습니다. 

따라서 우리는 노드가 취할 수 있는 최소값과 최대값을 추적해야 합니다. 그리고 각 노드에서 해당 값이 허용되는 최소값과 최대값 사이에 있는지 확인합니다. 루트는 음의 무한대와 양의 무한대 사이의 값을 가질 수 있습니다. 어떤 노드에서든 왼쪽 자식은 자신의 값보다 작거나 같아야 하며, 마찬가지로 오른쪽 자식도 크거나 같아야 합니다. 따라서 재귀 중에 현재 값을 새로운 최대값으로 왼쪽 자식에게 보내고 최소값은 변경하지 않고 있는 그대로 보냅니다. 그리고 오른쪽 자식에게는 현재 값을 새로운 최소값으로 보내고, 변경하지 않고 최대값을 보냅니다.

```python
class Node:
    def __init__(self, val=None):
        self.left, self.right, self.val = None, None, val


INFINITY = float("infinity")
NEG_INFINITY = float("-infinity")


def isBST(tree, minVal=NEG_INFINITY, maxVal=INFINITY):
    if tree is None:
        return True
    if not minVal <= tree.val <= maxVal:
        return False

    return isBST(tree.left, minVal, tree.val) and isBST(tree.right, tree.val, maxVal)
```

똑같이 좋은 대체 솔루션이 있습니다. 트리가 이진 검색 트리인 경우 트리 순서를 순회하면 트리 값이 정렬된 순서로 표시되어야 합니다. 따라서 중위 순회를 수행하여 노드 값이 정렬되어 있는지 확인할 수 있습니다.

```python
def isBST2(tree, lastNode=[NEG_INFINITY]):

    if tree is None:
        return True

    if not isBST2(tree.left, lastNode):
        return False

    if tree.val < lastNode[0]:
        return False

    lastNode[0] = tree.val

    return isBST2(tree.right, lastNode)
```

이는 일반적인 인터뷰 문제로, 상대적으로 간단하지만 사소하지 않으며 누군가가 이진 검색 트리 및 트리 순회에 대한 지식을 가지고 있음을 보여줍니다.
