## 문제

목표 금액과 가능한 동전 액면가 목록을 제공하고, 동전 액면가를 사용하여 목표 금액을 변경하는 방법의 수를 반환하는 함수를 작성하세요**

## 요구사항

종이/연필에 작업 내용을 작성한 다음 솔루션을 코딩할 수 있는지 확인하세요

## 솔루션

이것은 고전적인 인터뷰 문제입니다. 너무 고전적이어서 이미 재귀 섹션에서 매우 유사한 문제를 본 적이 있을 것입니다! 여기에서 솔루션을 읽기 전에 먼저 해당 문제를 검토하십시오!

이 솔루션에서는 [상향식](https://en.wikipedia.org/wiki/Top-down_and_bottom-up_design) 알고리즘을 사용합니다.

* 각 코인을 반복하면서 arr[i - 코인]을 만드는 방법을 arr[i]에 추가합니다.
* 4를 만드는 두 가지 방법이 있고 현재 값이 3인 동전을 반복하고 있다면 7을 만드는 두 가지 방법이 있어야 합니다.
* 우리는 본질적으로 arr[i]를 만드는 방법의 수에 반복 중인 코인을 추가하고 있습니다.

```python
def solution(n, coins):

    # Set up our array for trakcing results
    arr = [1] + [0] * n

    for coin in coins:
        for i in range(coin, n + 1):
            arr[i] += arr[i - coin]

    if n == 0:
        return 0
    else:
        return arr[n]
```

```python
solution(100, [1, 2, 3])
```

이 솔루션의 결과는 O((m)(n))이며 m은 동전 수이며 여기서 n 작업을 반복합니다. 이것은 O(n) 공간입니다.
