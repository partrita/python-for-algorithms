# 계란 드롭 
이것은 아마도 이 그룹에서 가장 흔한 두뇌 자극 수수께끼일 것입니다. 그러니 해결책을 보기 전에 이 문제에 대해 알고리즘적으로 생각해보세요!
## 문제 설명

타워에는 100층이 있습니다. 당신은 계란 두 개를 받았습니다. 계란은 깨지지 않고 탑의 특정 층에서 떨어질 수 있을 만큼 강력합니다. 당신은 가능한 한 적은 수의 방울로 계란이 깨지지 않고 떨어질 수 있는 가장 높은 층을 찾는 임무를 받았습니다. 계란이 목표 층 위에서 떨어지면 계란이 깨집니다. 해당 층 이하에서 떨어뜨리면 그대로 유지되며 다른 층에 다시 시험적으로 달걀을 떨어뜨릴 수 있습니다.

가능한 한 적은 방울로 이 작업을 수행하는 방법을 알고리즘적으로 보여줍니다. (답은 100층에서 계란 2개를 테스트하는 데 필요한 최소 방울 수여야 합니다.)
