# 에그 드롭 - 솔루션
이것은 아마도 이 그룹에서 가장 흔한 두뇌 자극 수수께끼일 것입니다. 그러니 해결책을 보기 전에 이 문제에 대해 알고리즘적으로 생각해보세요!

## 문제 설명

타워에는 100층이 있습니다. 당신은 계란 두 개를 받았습니다. 계란은 깨지지 않고 탑의 특정 층에서 떨어질 수 있을 만큼 강력합니다. 당신은 가능한 한 적은 수의 방울로 계란이 깨지지 않고 떨어질 수 있는 가장 높은 층을 찾는 임무를 받았습니다. 계란이 목표 층 위에서 떨어지면 계란이 깨집니다. 해당 층 이하에서 떨어뜨리면 그대로 유지되며 다른 층에 다시 시험적으로 달걀을 떨어뜨릴 수 있습니다.

가능한 한 적은 방울로 이 작업을 수행하는 방법을 알고리즘적으로 보여줍니다. (답은 100층에서 계란 2개를 테스트하는 데 필요한 최소 방울 수여야 합니다.)
## 솔루션

10층부터 시작해서 10의 배수로 층수까지 올라갑니다.

첫 번째 계란이 깨지면, 예를 들어 20층에서 두 번째 계란으로 11층과 19층 사이의 모든 층을 확인하여 어느 층이 깨지지 않는지 확인할 수 있습니다.

이 경우 최악의 낙하 횟수는 19입니다. 임계값이 99층이라면 선형 방식으로 첫 번째 알을 10번, 두 번째 알을 9번 떨어뜨려야 합니다.

**최상의 솔루션:**
우리는 최악의 낙하 횟수를 최소화해야 합니다. 이를 위해서는 문제를 n개 층으로 일반화해야 합니다. 첫 번째 계란의 단계 값은 얼마입니까? 그래도 10이겠지? 200층이 있다고 가정해 봅시다. 단계 값이 여전히 10인가요? 

여기서 주목해야 할 점은 임계값이 가장 높은 층에 있을 때 발생하는 최악의 경우 드롭 횟수를 최소화하려고 노력하고 있다는 것입니다. 따라서 우리의 단계는 첫 번째 계란의 방울 수를 줄이는 어느 정도 가치가 있어야 합니다.

처음에 어떤 단계 값 m을 취한다고 가정해 봅시다. 모든 후속 단계가 m-1이면,
그럼, 
$$m+m−1+m−2+.....+1=n$$

이것은 

$$\frac{m*(m+1)}{2}=n$$

n =100이면 m은 13.65가 됩니다. 소수점 아래로 내려갈 수 없기 때문에 실제로는 14를 사용합니다.

따라서 최악의 시나리오는 이제 임계값이 처음 14층에 있고 드롭 수가 14개인 경우입니다.

이는 단순한 이진 검색이라는 점에 유의하세요!
