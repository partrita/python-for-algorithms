# 동전과 저울 - 솔루션
## 문제 설명

동전 8개와 팬 저울이 있습니다. 모든 동전의 무게는 동일합니다. 단, 다른 동전보다 무거운한 동전**만 제외하면 됩니다. 그렇지 않으면 동전을 구별할 수 없습니다. 무거운 동전이 얼마나 무거운지에 대해 아무런 가정도 할 수 없습니다. 무거운 동전을 확실히 식별하는 데 필요한 최소 무게 측정 횟수는 얼마입니까?

## 솔루션

동전을 3개씩 두 그룹으로 나누어 저울에 올리고, 2개씩 한 그룹은 놔두세요. 양쪽의 무게가 같다면 무거운 동전은 2개의 그룹에 속하며, 한 번 더 달아서 총 2번의 무게로 찾을 수 있습니다. 반면, 저울의 어느 한 쪽이 더 무거우면 무거운 동전은 세 개 그룹에 속해야 합니다. 다른 동전을 모두 제거하고 이 그룹의 동전 하나를 저울 양쪽에 놓고 세 번째 동전은 따로 남겨 둘 수 있습니다. 한쪽이 무거우면 무거운 동전이 들어갑니다. 어느 쪽도 무겁지 않다면, 저울에 올리지 않은 동전이 무거운 동전입니다. 이것도 총 2번의 저울이기 때문에 2번의 저울을 이용하면 항상 8개의 그룹에서 무거운 동전을 찾을 수 있습니다.
