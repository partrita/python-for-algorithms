# 다리 건너기 - 솔루션

## 문제 설명

네 명의 여행자 그룹이 밤에 다리에 왔습니다. 다리는 한 번에 최대 두 명의 여행자만 지탱할 수 있으며, 손전등을 사용하지 않으면 건널 수 없습니다. 

여행자 중에는 손전등이 하나 있습니다. 여행자마다 걷는 속도는 다릅니다. 첫 번째 사람은 1분 만에 다리를 건널 수 있고, 두 번째 사람은 2분 만에, 세 번째 사람은 5분 만에, 네 번째 사람은 다리를 건너는 데 10분이 걸립니다. 두 명의 여행자가 함께 건너면 느린 여행자의 속도로 걷습니다.

모든 여행자가 다리의 한쪽에서 다른 쪽으로 건너갈 수 있는 최소 시간은 얼마입니까?

## 솔루션

이것은 [강 건너기](https://en.wikipedia.org/wiki/River_crossing_puzzle) 퍼즐의 일반적인 그룹의 일부입니다. 이는 [브리지 및 횃불 문제](https://en.wikipedia.org/wiki/Bridge_and_torch_problem)로 알려져 있습니다(때때로 사람마다 할당된 시간이 다릅니다).
