# 복도 사물함 - 솔루션
## 문제 설명

당신은 100개의 사물함이 늘어선 복도에 있습니다. 한 번의 패스로 시작하여 사물함을 열면 이제 *열린* 사물함의 문이 열린 상태가 됩니다.매초사물함을 닫는 것부터 시작합니다. 그런 다음세 번째마다사물함을 닫고열려 있으면 닫고, 닫혀 있으면 엽니다— 이것을 사물함 "토글"이라고 합니다. 패스 번호 n에서 매 n번째 사물함을 계속 전환합니다. 100번 복도를 통과한 후 사물함 번호 100만 전환하면 열려 있는 사물함은 몇 개입니까?

## 솔루션

분명히 무차별 대입으로 이 문제를 계산할 수는 없습니다. 패스가 너무 많기 때문에 알고리즘적으로 이 문제를 생각해야 합니다.

임의의 사물함을 선택하여 이 문제를 해결하고 패턴을 감지할 수 있는지 살펴보겠습니다. 12번 사물함을 선택하세요. 첫 번째 패스에서 열리도록 전환되어 있습니다.

시작하려면 12보다 큰 패스에서는 이를 토글할 필요가 없다는 것을 알고 있습니다. 따라서 이제 2-11에서 발생하는 패스만 생각하면 됩니다. 우리는 실제로 다음을 셀 수 있습니다:

* 패스 2: 2,4,6,8,10,12
* 패스 3: 3,6,9,12
* 패스 4: 4,8,12
* 패스 5: 5,10이 패스에는 토글이 없습니다**
* 패스 6: 6,12
* 패스 7: 7,14이 패스에는 토글이 없습니다**
* 등등...

나타나는 패턴을 볼 수 있을 것입니다. 패스 번호가 라커 번호의 일부인 경우에만 라커를 전환합니다. 우리는 첫 번째 패스 이후 모든 라커가 열리기 시작했고 열림과 닫힘 사이를 번갈아가며 일반화하기 시작할 수 있습니다. 따라서 보관함은 두 번째, 네 번째, 여섯 번째 등으로 전환된 후에 닫힙니다. 즉, 보관함이 짝수 번 전환되면 닫히게 됩니다. 그렇지 않으면 열린 채로 끝납니다. 사물함은 사물함 번호의 모든 요소에 대해 한 번씩 전환된다는 것을 알고 있으므로 사물함의 요소 수가 홀수인 경우에만 사물함이 열린다고 말할 수 있습니다.

**이제 작업은 1에서 100 사이에서 홀수 개의 요인이 있는 숫자를 찾는 것으로 축소되었습니다!**
____

우리는 이에 대해 다음과 같은 방식으로 생각할 수 있습니다.

숫자i**가n**의 인수라면 이는 무엇을 의미합니까? 이는i곱하기 다른 숫자j**가 n과 같다는 것을 의미합니다. 곱셈은 ​​교환 가능하므로(i × j = j × i), 이는 j도 n의 인수임을 의미하며, 인수가 쌍으로 오는 경향이 있기 때문에 인수의 수는 일반적으로 짝수입니다. 짝을 이루지 않은 요소가 있는 번호를 찾을 수 있으면 어떤 사물함이 열려 있는지 알 수 있습니다. 곱셈은 ​​이항 연산이므로 항상 두 개의 숫자가 관련됩니다. 하지만 둘 다 같은 숫자(즉, i = j)이면 어떻게 될까요? 이 경우 단일 숫자가 효과적으로 쌍의 양쪽 절반을 형성하게 되며 홀수 개의 요인이 있게 됩니다. 이 경우 i × i = n이 됩니다. 따라서 n은 완전제곱수여야 합니다. 이 해법을 확인하려면 완전제곱식을 시도해 보세요. 예를 들어,
16의 경우 요소는 1, 2, 4, 8, 16입니다. 작업은 open, close, open, close, open입니다. 예상대로 open으로 끝납니다.

이러한 추론을 바탕으로 숫자가 완전제곱수인 사물함만 열린다는 결론을 내릴 수 있습니다. 1과 100(포함) 사이의 완전제곱수는 1, 4, 9, 16, 25, 36, 49, 64, 81, 100입니다. 따라서 10개의 사물함은 열려 있습니다.
