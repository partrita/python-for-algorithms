# 전등 스위치 - 솔루션
## 문제 설명

당신은 세 개의 전등 스위치 옆 복도에 있는데, 모두 꺼져 있습니다. 각 스위치는 복도 끝에 있는 방에 있는 서로 다른 *백열* 전구를 활성화합니다. 스위치가 있는 곳에서는 조명을 볼 수 없습니다. 당신의 임무는 각 스위치에 해당하는 조명을 결정하는 것입니다. 단, 불이 켜진 방에는 한 번만 들어갈 수 있습니다.

**참고: 이것은 "트릭" 질문이므로 너무 많은 시간을 투자하지 마십시오. 수수께끼 유형의 질문 중 "재미있는" 측면에 더 가깝지만.**

## 솔루션

이는 다소 까다로운 질문이므로 인터뷰에서 이런 유형의 질문을 받지 않기를 바랍니다. 왜냐하면 실제로는 수학이나 논리에 기반한 질문이 아니기 때문입니다. 해결책은 *백열등* 전구가 뜨거워질 때까지 잠시 켜둘 수 있다는 사실을 깨닫는 것입니다. 

따라서 해결책은 스위치 1을 켜고 해당 전구가 뜨거워질 때까지 15분 동안 기다리는 것입니다. 그런 다음 전원을 끄고 스위치 2를 켠 다음 방으로 향합니다. 그러면 다음 사항을 알 수 있습니다.

* 뜨거운 전구는 스위치 1에 해당합니다.
* 켜져 있는 전구는 스위치 2에 해당합니다.
* 꺼진 전구는 스위치 3에 해당합니다.
