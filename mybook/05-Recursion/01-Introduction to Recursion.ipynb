## 재귀란 무엇인가요?

재귀에는 두 가지 주요 사례가 있습니다. 첫 번째는 함수가 자신을 하나 이상 호출하는 기술로 재귀를 사용하는 경우입니다. 두 번째는 데이터 구조가 자신을 나타낼 때 정확히 동일한 유형의 데이터 구조의 더 작은 인스턴스를 사용하는 경우입니다. 이 두 인스턴스는 모두 재귀의 사용 사례입니다.

식물에서 볼 수 있는 프랙탈 패턴과 같이 현실 세계에서도 실제로 재귀가 발생합니다!

## 왜 재귀를 사용하나요?

재귀는 루프가 이상적이지 않은 작업을 반복 수행하기 위한 강력한 대안을 제공합니다. 대부분의 최신 프로그래밍 언어는 재귀를 지원하며 재귀는 특정 데이터 구조를 구축하는 데 훌륭한 도구 역할을 합니다.

# 계승 예

이번 강의에서는 계승함수 생성 연습을 통해 재귀에 대해 설명하겠습니다.
계승 함수는 느낌표로 표시되며 1부터 *n*까지의 정수의 곱으로 정의됩니다. 공식적으로는 다음과 같이 말할 수 있습니다.

$$ 엔! = n·(n-1)·(n-2)... 3·2·1 $$

**n = 0이면 n! = 1**. 이는 *기본 사례* 역할을 하기 때문에 고려하는 것이 중요합니다.

다음 예를 들어보세요.
$$4! = 4 · 3 · 2 · 1 = 24. $$
그렇다면 이것을 재귀적인 방식으로 어떻게 표현할 수 있을까요? 여기서기본 사례개념이 등장합니다.

**기본 사례**는 특히 재귀를 다루는 인터뷰 문제를 해결해야 하는 경우 재귀를 이해하는 데 중요한 부분입니다. 위 4의 방정식을 다시 써보자! 그래서 그것은 다음과 같습니다:

$$ 4! = 4 · (3 · 2 · 1) = 24 $$

이는 다음과 동일합니다.

$$ 4! = 4 · 3! = 24 $$

즉, 다음과 같이 재귀 측면에서 형식적인 재귀 정의를 다시 작성할 수 있습니다.

$$ 엔! = n·(n−1)!$$

**n = 0이면 n! = 1**. 이는 n=0일 때기본 사례**가 발생하고 *재귀 사례*가 위 방정식에 정의됨을 의미합니다. 재귀 솔루션을 개발하려고 할 때마다 기본 사례에 대해 생각하는 것이 매우 중요합니다. 모든 재귀 사례가 처리되면 솔루션이 기본 사례를 반환해야 하기 때문입니다. Python에서 계승 함수를 생성하는 방법을 살펴보겠습니다.
___

```python
def fact(n):
    """
    Returns factorial of n (n!).
    Note use of recursion
    """
    # BASE CASE!
    if n == 0:
        return 1

    # Recursion!
    else:
        return n * fact(n - 1)
```

실제로 확인해 봅시다!

```python
fact(5)
```
