# 문자열 순열

## 문제 설명

문자열이 주어지면 재귀를 사용하여 해당 문자열의 가능한 모든 순열 목록을 출력하는 함수를 작성합니다.

예를 들어, s='abc'인 경우 함수는 ['abc', 'acb', 'bac', 'bca', 'cab', 'cba']를 반환해야 합니다.

*참고: 문자가 반복되는 경우 각 발생을 별개로 처리하십시오. 예를 들어 'xxx'를 입력하면 'xxx'의 6개 "버전"이 포함된 목록이 반환됩니다.*


## 아래에서 솔루션을 작성하세요.

여기서 우리가 취해야 할 단계는 무엇인지 생각해 봅시다.

1. 초기 문자열(예: 'abc')을 반복합니다.

* 초기 문자열의 각 문자에 대해 해당 문자를 따로 보관하고 남은 문자열의 모든 순열 목록을 가져옵니다. 예를 들어 현재 반복이 'b'에 있는 경우 문자열 'ac'의 모든 순열을 찾고 싶을 것입니다.

* 2단계의 목록이 있으면 해당 목록의 각 요소를 초기 문자열의 문자에 추가하고 결과를 최종 결과 목록에 추가합니다. 따라서 우리가 'b'에 있고 목록 ['ac', 'ca']를 얻은 경우 여기에 'b'를 추가하여 'bac'와 'bca'가 되고 각각을 최종 결과에 추가합니다.

* 최종 결과 목록을 반환합니다.

계속해서 이것이 구현되었는지 살펴보겠습니다.

```python
def permute(s):
    out = []

    # Base Case
    if len(s) == 1:
        out = [s]

    else:
        # For every letter in string
        for i, let in enumerate(s):
            # For every permutation resulting from Step 2 and 3 described above
            for perm in permute(s[:i] + s[i + 1 :]):
                # Add it to output
                out += [let + perm]

    return out
```

```python
permute("abc")
```

# 솔루션 테스트

```python
"""
RUN THIS CELL TO TEST YOUR SOLUTION.
"""

from nose.tools import assert_equal


class TestPerm(object):
    def test(self, solution):

        assert_equal(
            sorted(solution("abc")), sorted(["abc", "acb", "bac", "bca", "cab", "cba"])
        )
        assert_equal(
            sorted(solution("dog")), sorted(["dog", "dgo", "odg", "ogd", "gdo", "god"])
        )

        print("All test cases passed.")


# Run Tests
t = TestPerm()
t.test(permute)
```
