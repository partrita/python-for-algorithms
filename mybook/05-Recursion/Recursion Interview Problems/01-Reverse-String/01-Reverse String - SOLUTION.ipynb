## 솔루션

재귀를 사용하여 문자열을 뒤집으려면 기본 케이스와 재귀 케이스가 어떻게 생겼는지 고려해야 합니다. 여기서는 함수를 통해 전달하는 문자열의 길이가 1보다 작거나 같은 기본 사례를 설정했습니다.

재귀적인 경우에는 첫 번째 문자를 가져와 재귀 호출에 추가합니다.

```python
def reverse(s):

    # Base Case
    if len(s) <= 1:
        return s

    # Recursion
    return reverse(s[1:]) + s[0]
```

```python
reverse("hello world")
```
