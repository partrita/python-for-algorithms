# 피보나치 수열

## 문제 설명

세 가지 방법으로 [피보나치 수열](https://en.wikipedia.org/wiki/Fibonacci_number)을 구현하세요.

* 재귀적으로
* 동적으로(메모이제이션을 사용하여 결과 저장)
* 반복적으로

피보나치 수열: 0,1,1,2,3,5,8,13,21,...은 n = 0인지 1인지 확인하는 기본 사례로 시작한 다음 1을 반환한다는 점을 기억하세요. 

그렇지 않으면 fib(n-1)+fib(n+2)를 반환합니다.

### 재귀적으로

재귀 솔루션은 O(2^n)을 사용하는 지수 시간 Big-O입니다. 그러나 고려해야 할 매우 간단하고 기본적인 구현은 다음과 같습니다.

```python
def fib_rec(n):

    # Base Case
    if n == 0 or n == 1:
        return n

    # Recursion
    else:
        return fib_rec(n - 1) + fib_rec(n - 2)
```

```python
fib_rec(10)
```

### 동적으로

여기에 구현된 형식에서는 캐시가 미리 설정되고 원하는 피보나치 수열의n수를 기반으로 합니다. 캐시를 어떻게 확인하는지 주목하세요.

```python
# Instantiate Cache information
n = 10
cache = [None] * (n + 1)


def fib_dyn(n):

    # Base Case
    if n == 0 or n == 1:
        return n

    # Check cache
    if cache[n] != None:
        return cache[n]

    # Keep setting cache
    cache[n] = fib_dyn(n - 1) + fib_dyn(n - 2)

    return cache[n]
```

```python
fib_dyn(10)
```

### 반복적으로

이 솔루션에서는 Python의 튜플 압축 풀기 기능을 활용할 수 있습니다!

```python
def fib_iter(n):

    # Set starting point
    a = 0
    b = 1

    # Follow algorithm
    for i in range(n):
        a, b = b, a + b

    return a
```

```python
fib_iter(23)
```
