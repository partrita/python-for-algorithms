# 재귀 숙제 문제 

이 과제는 재귀 개념에 익숙해지기 시작하는 다양한 작은 문제입니다. 본격적인 인터뷰 질문은 아니지만 재귀 문제에 대한 "상태"에 마음을 두는 데 훌륭한 시작 역할을 합니다.


______
### 문제 1

**정수를 취하고 그 정수에 대한 0의 누적합을 계산하는 재귀 함수를 작성하세요**

**예를 들어 n=4인 경우 4+3+2+1+0, 즉 10을 반환합니다.**

이 문제는 재귀를 소개하는 동안 제시된 계승 문제와 매우 유사합니다. 기본 사례가 어떻게 보일지 항상 생각하십시오. 이 경우 n =0의 기본 사례가 있습니다(참고: 컷오프를 1로 설계할 수도 있음).

이 경우 다음이 있습니다.
   n + (n-1) + (n-2) + .... + 0

샘플 솔루션을 작성하세요.

```python
def rec_sum(n):
    if n == 0:
        return 0
    else:
        return n + rec_sum(n - 1)
    pass
```

```python
rec_sum(4)
```

### 문제 2

**정수가 주어지면 해당 정수의 모든 개별 숫자의 합을 반환하는 함수를 만듭니다. 예를 들면:
n = 4321이면 4+3+2+1을 반환**

```python
def sum_func(n):
    if n < 10:
        return n
    else:
        return n % 10 + sum_func(int(n / 10))
    pass
```

```python
sum_func(4321)
```

### 문제 3
*참고로 이 문제는 이전 두 문제보다 더 발전된 문제입니다! 또한 다양한 변형 가능성이 있으므로 여기서는 엄격한 요구 사항을 무시합니다.*

문자열phrase**와list_of_words집합을 받는 word_split()라는 함수를 만듭니다. 그런 다음 함수는 단어 목록에서 단어를 만들 수 있는 방식으로 문자열을 분할할 수 있는지 확인합니다. 구문이 완전히 분할 가능한 경우 사전에 있는 단어만 포함한다고 가정할 수 있습니다.

예를 들어:

```python
word_split("themanran", ["the", "ran", "man"])
```

```python
word_split("ilovedogsJohn", ["i", "am", "a", "dogs", "lover", "love", "John"])
```

```python
word_split("themanran", ["clown", "ran", "man"])
```

```python
def word_split(phrase, list_of_words, output=None):
    if output is None:
        output = []

    for word in list_of_words:
        if phrase.startswith(word):
            output.append(word)
            word_split(phrase[len(word) :], list_of_words, output)
    return output
    pass
```
