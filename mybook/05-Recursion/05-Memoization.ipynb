# 메모하기

이번 강의에서는 메모이제이션(Memoization)과 다이나믹 프로그래밍(Dynamic 프로그래밍)에 대해 알아보겠습니다. 숙제는 [메모에 관한 위키피디아 글](https://en.wikipedia.org/wiki/Memoization)을 읽어보시고 강의를 진행해주세요!

____

메모이제이션이란 메서드 입력을 기반으로 메서드 호출 결과를 기억("메모" -> "메모" -> 기억되도록)한 다음, 결과를 다시 계산하는 것이 아니라 기억된 결과를 반환하는 것을 효과적으로 의미합니다. 메소드 결과에 대한 캐시로 생각할 수 있습니다. 우리는 이것을 순전히 재귀적인 솔루션의 개선된 버전으로 일부 인터뷰 문제에서 사용할 것입니다.

Python에서 메모이제이션을 사용하여 계승을 계산하는 간단한 예는 다음과 같습니다.

```python
# Create cache for known results
factorial_memo = {}


def factorial(k):

    if k < 2:
        return 1

    if k not in factorial_memo:
        factorial_memo[k] = k * factorial(k - 1)

    return factorial_memo[k]
```

```python
factorial(4)
```

이제 팩토리얼 함수의 이전 결과를 저장하기 위해 어떻게 사전을 사용하고 있는지 주목하세요! 이제 이전 결과를 기억하여 이 기능의 효율성을 높일 수 있습니다!

메모 처리 프로세스를 클래스로 캡슐화할 수도 있습니다.

```python
class Memoize:
    def __init__(self, f):
        self.f = f
        self.memo = {}

    def __call__(self, *args):
        if args not in self.memo:
            self.memo[args] = self.f(*args)
        return self.memo[args]
```

그러면 우리가 해야 할 일은 다음과 같습니다.

```python
def factorial(k):

    if k < 2:
        return 1

    return k * factorial(k - 1)


factorial = Memoize(factorial)
```
